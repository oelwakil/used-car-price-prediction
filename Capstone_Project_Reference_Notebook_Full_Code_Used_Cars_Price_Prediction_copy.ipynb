# **Cars4U Used Car Price Prediction Project**

## **Executive Summary: Optimizing Used Car Pricing for Cars4U**

### **Problem and Strategic Imperative:**
Cars4U operates in a dynamic Indian used car market characterized by significant pricing volatility. Inaccurate pricing has directly impacted our profitability, leading to lost revenue from underpriced vehicles and reduced sales velocity from overpriced inventory. Our strategic objective was to deploy a data-driven model that accurately estimates used car market prices, ensuring competitive positioning and enhanced financial performance.

### **Key Insights and Model Development Successes:**

Our comprehensive analysis and model development yielded several critical insights and a highly effective predictive solution:

1.  **New Car Price is Paramount:** The single most influential factor in determining a used car's value is its original new price. Models leveraging this data achieved significantly higher predictive accuracy. This underscores the strategic importance of systematically acquiring new car pricing information.
2.  **Depreciation Dynamics Quantified:** We successfully mapped the complex interplay of age, mileage, and engine characteristics on car depreciation. Our model accurately captures how combined factors, such as high mileage on an older car, accelerate value decline, providing granular insights into true market value.
3.  **Market Nuances Captured:** Regional market variations, brand prestige, fuel type, transmission, and ownership history significantly influence used car prices. Our model integrates these nuanced factors, demonstrating that location-specific demand, brand perception, and vehicle specifications play a crucial role beyond basic depreciation.
4.  **Advanced Analytics Deliver Superior Accuracy:** Leveraging a tuned **XGBoost Regressor model**, we achieved exceptional predictive performance. The model explains over **97% of the variation** in used car prices (Test R-squared: 0.9709) and predicts prices with an average error of just **6.10% (Test MAPE)**. This represents a substantial improvement over traditional statistical methods, directly translating to more precise and profitable pricing.
5.  **Interpretability Unlocked:** While sophisticated, the model's decisions are transparent. Using SHAP analysis, we can explain *why* any specific car received its predicted price, identifying which features pushed the price up or down. This builds trust and provides actionable insights for sales and purchasing teams.

### **Final Proposed Model Specifications:**
*   **Model Type:** XGBoost Regressor (Optuna Tuned) – a robust, high-performance ensemble learning algorithm.
*   **Key Predictive Features:** Includes original new car price, engine power (via a combined PCA component), car age, mileage, and critical interaction terms capturing depreciation dynamics. It also accounts for geographical location, brand, fuel type, transmission, and owner history.
*   **Data Foundation:** Operates on a meticulously cleaned and engineered dataset of approximately 2,600 car listings with complete new price information, ensuring high data quality for optimal predictions.

### **Key Next Steps and Strategic Recommendations for Cars4U:**

1.  **Immediate Deployment for Business Impact:** Integrate the tuned XGBoost model into Cars4U's operational pricing systems. This will immediately enable more accurate pricing decisions, directly contributing to increased revenue, improved inventory turnover, and enhanced market competitiveness.
2.  **Prioritize New Car Price Data Acquisition:** Launch a dedicated initiative to secure comprehensive and accurate new car pricing data across all models. Given its profound impact on model accuracy, this is our most critical data strategy to maintain and further enhance predictive power.
3.  **Establish Continuous Model Monitoring and Refinement:** Implement an MLOps framework for ongoing performance monitoring and regular model retraining. The dynamic nature of the used car market necessitates an adaptive model that evolves with new data and changing market conditions.
4.  **Empower Sales Teams with Interpretability:** Train sales and purchasing teams on using SHAP explanations to understand and articulate model-derived prices. This will build confidence in our pricing strategy and support effective negotiation with both sellers and buyers.
5.  **Explore Strategic Enhancements:** Post-deployment, investigate opportunities for further refinement, such as developing segment-specific models for luxury or electric vehicles, or integrating external market indicators to capture broader economic trends.

This robust, data-driven approach promises to transform Cars4U's pricing strategy, solidifying our market position and driving sustainable growth.

## **Problem and Solution Summary**

### **Problem Being Solved**

Cars4U faced a critical challenge in the fragmented and volatile Indian used car market: inaccurate pricing. This led to **overpricing (reducing sales velocity)** and **underpricing (resulting in revenue loss)**, directly impacting customer trust, inventory turnover, and overall profitability. The objective was to develop a data-driven predictive model to accurately estimate used car market prices.

### **Reason for the Proposed Solution Design**

1.  **High Predictive Accuracy:** The chosen solution, the **XGBoost Regressor (Optuna Tuned) model**, provides exceptional predictive power. It achieved a Test R-squared of 0.9704 and a Test MAPE of 6.19%, significantly outperforming traditional linear models. This high accuracy directly addresses the core problem of inaccurate pricing.
2.  **Capturing Complex Relationships:** Used car pricing is influenced by intricate, non-linear relationships and interactions between various features (e.g., how age impacts depreciation differently for luxury versus economy cars, or the combined effect of mileage and age). As an ensemble tree-based model, XGBoost inherently excels at capturing these complexities, which linear models, even with transformations, struggle to model effectively.
3.  **Robustness:** XGBoost is robust to multicollinearity (a significant issue identified among several features, especially after one-hot encoding) and outliers. Its architecture makes it less sensitive to these data imperfections compared to OLS, leading to more stable and reliable predictions.
4.  **Enhanced Interpretability (with SHAP):** While often considered a 'black box', the use of SHAP (SHapley Additive exPlanations) analysis provides valuable interpretability for the XGBoost model. SHAP reveals the directional impact and magnitude of each feature's contribution to individual predictions, offering transparent insights into *why* a car received a certain price. This addresses the need for understanding and trust in the model's outputs.
5.  **Leveraging Critical Data:** The model effectively utilizes the highly influential `log_New_price` feature and its interactions, which proved indispensable for achieving top-tier accuracy. This demonstrates the value of strategic data inclusion despite initial missingness challenges.

### **How it would Affect the Problem/Business & Why it's a Valid Solution**

The implementation of this XGBoost model would profoundly affect Cars4U by:
*   **Optimizing Revenue:** By providing highly accurate price estimations, the model minimizes underpricing (preventing revenue loss) and overpricing (improving sales velocity and reducing inventory holding costs).
*   **Boosting Customer Trust:** Transparent and data-driven pricing fosters greater customer confidence, leading to a better reputation and potentially higher conversion rates.
*   **Enabling Data-Driven Decisions:** It allows Cars4U to make informed, objective pricing decisions, moving away from subjective human judgment alone, thereby enhancing competitive advantage.
*   **Scaling Operations:** An automated, accurate pricing mechanism supports efficient scaling of operations across different locations and vehicle types.

This is a 'valid' solution because its demonstrated high accuracy (low RMSE, MAE, and MAPE) on unseen test data, coupled with its ability to handle real-world data complexities, ensures that it will consistently provide reliable price predictions. The interpretability provided by SHAP further validates its utility by offering actionable insights into pricing drivers.

# **Used Cars Price Prediction**

## **Problem Definition**

### **The Context:**

 - Why is this problem important to solve?

 The Indian automobile market has undergone a significant shift in recent years, with the used car market surpassing the new car market in volume. In 2018–19 alone, approximately 4 million used cars were sold compared to 3.6 million new cars, highlighting a clear change in consumer behavior. Rising new car prices, economic uncertainty, and increased trust in certified pre-owned vehicles have accelerated this trend.

However, unlike the new car market, where prices are relatively standardized and controlled by manufacturers, the used car market is highly fragmented and volatile. Prices vary significantly based on multiple factors such as brand, age, mileage, location, fuel type, and ownership history. This makes pricing a used car both complex and subjective, often leading to:

*   **Overpricing, which reduces sales velocity**
*   **Underpricing, which leads to revenue loss**

**For a platform like Cars4U, inaccurate pricing directly impacts customer trust, inventory turnover, and profitability. Therefore, building a data-driven pricing model becomes crucial to ensure competitive, transparent, and profitable pricing decisions.**

### **The objective:**

 - What is the intended goal?

 ***The primary goal of this project is to develop a predictive model that can accurately estimate the market price of a used car based on its attributes.***

Specific objectives include:

* **Predicting used car prices with high accuracy using historical data.**

* **Identifying key factors that significantly influence a car’s resale value.**

* **Reducing pricing inconsistencies across listings.**

* **Enabling data-driven decision-making for pricing strategies.**

* **Supporting business growth by improving customer trust and transaction efficiency.**

### **The key questions:**

- What are the key questions that need to be answered?

To effectively solve the problem, the following critical questions need to be addressed:

* ***Which features (e.g., mileage, age, brand, fuel type) have the greatest impact on used car prices?***

* ***How does the price of a car depreciate with age and usage?***

* ***Do location and transmission type significantly affect resale value?***

* ***Can we accurately predict a car’s selling price using historical data?***

* ***How much variance in price can be explained by the available features?***

* ***Are there patterns or segments (e.g., luxury vs economy cars) that behave differently in terms of pricing?***

These questions guide both exploratory analysis and model development.

### **The problem formulation**:

- What is it that we are trying to solve using data science?

**From a data science standpoint, this problem can be formulated as a supervised regression problem, where:**

* **Input variables (features):**
Car attributes such as year, mileage, fuel type, transmission, engine capacity, power, location, ownership type, etc.

* **Target variable:**
The used car price (continuous numerical value).

**The goal is to learn a function that maps car features to its expected market price using historical data.**

**Role of Data Science in Solving the Problem:**

* **Data Cleaning & Preprocessing:** Handle missing values, outliers, and inconsistent units.

* **Exploratory Data Analysis (EDA):** Understand relationships and trends among variables.

* **Feature Engineering:** Transform and select features that improve predictive power.

* **Model Building:** Apply regression-based machine learning algorithms (e.g., Linear Regression, Random Forest, Gradient Boosting).

* **Model Evaluation:** Use metrics such as RMSE, MAE, and R² to assess performance.

* **Business Interpretation:** Translate model insights into actionable pricing strategies.

### **Data Dictionary**

**S.No.** : Serial Number

**Name** : Name of the car which includes Brand name and Model name

**Location** : The location in which the car is being sold or is available for purchase (Cities)

**Year** : Manufacturing year of the car

**Kilometers_driven** : The total kilometers driven in the car by the previous owner(s) in KM

**Fuel_Type** : The type of fuel used by the car (Petrol, Diesel, Electric, CNG, LPG)

**Transmission** : The type of transmission used by the car (Automatic / Manual)

**Owner** : Type of ownership

**Mileage** : The standard mileage offered by the car company in kmpl or km/kg

**Engine** : The displacement volume of the engine in CC

**Power** : The maximum power of the engine in bhp

**Seats** : The number of seats in the car

**New_Price** : The price of a new car of the same model in INR 100,000

**Price** : The price of the used car in INR 100,000 (**Target Variable**)

**Addressing Missing Values of Price (Target Variable):**

For the 'Price' column (target variable), the standard and most appropriate approach is to **drop the rows where 'Price' is missing** for the following reasons:
  * **Model Training Requirement:** Supervised machine learning models need both input features and a known output feature to learn the relationship between them. If a row has a missing 'Price', there's no ground truth for the model to learn from.
  * **Avoiding Bias from Imputation:** Imputing the target variable could introduce significant bias. If the missing prices are imputed, the model would learn from these guesses, leading to an artificially inflated performance that might not reflect its true predictive capability on unseen, real-world data.
  * **Preserving Data Integrity:** Dropping rows with missing target values ensures that our model is trained only on complete and reliable information regarding the outcome that is being predicted.

**Dropping Rows with Missing Values of Price (Target Variable):**

````python
# Dropping Rows with Missing Values of Price (Target Variable)
df.dropna(subset=['Price'], inplace=True)

# Displaying info of updated data
df.info()
````

**Strategy for Missing Values of Features:**

| Feature   | Missing | % Missing |
| :-------- | :------ | :-------- |
| New_price | 5195    | 86.31%    |
| Power     | 143     | 2.38%     |
| Seats     | 42      | 0.70%     |

* **Step 1 - Handle missing values in 'Seats', 'Engine', and 'Mileage', while preserving 'Electric' cars:**

  * **Impute 'Mileage' for 'Electric' cars:** The 2 missing 'Mileage' values correspond to 'Electric' cars. To preserve these important observations, their 'Mileage' will be imputed to 1.0. This value, when log-transformed, will become 0, acting as a neutral placeholder that signifies 'no fuel consumption' without distorting the mileage distribution for fuel-burning cars.
  * **Drop remaining rows with missing 'Seats' and 'Engine':** After imputing 'Mileage' for 'Electric' cars, the remaining rows with missing 'Seats' (42) and 'Engine' (36) will be dropped. These are crucial structural attributes, and their reliable imputation is difficult. The data loss (less than 80 rows) is negligible and ensures data integrity.

* **Step 2 - Impute missing values of 'Power' using 'Brand' from 'Name', 'Year', 'Fuel_Type','Transmission', 'Engine':**

  * 'Power' is highly structured and strongly determined by engine configuration and model. Imputing it based on these related features is a reasonable approach.

* **Step 3 - Drop remaining rows with missing values of 'Power':**

  * Any 'Power' values that could not be imputed using the grouped median method are likely from rare configurations or poorly specified listings. Dropping these few remaining rows avoids introducing noisy imputations.

* **Step 4 - Impute 'New_price' using 'Brand', 'Year','Fuel_Type', 'Transmission', 'Engine', 'Power', and 'Seats' (in a phased approach):**

  * The 'New_price' column has a very high percentage of missing values. While direct imputation based on closely related features (Brand, Year, etc.) is the best approach, a phased imputation strategy will be employed. An initial granular imputation (including 'Power') will capture specific matches, followed by a broader imputation (excluding 'Power') to maximize coverage, acknowledging that noise introduced in 'Power' imputation can propagate but is acceptable given 'New_price' is not the target and the primary predictors are stronger.

### **Missing Values Treatment**

**Step 1 - Handling missing values in 'Seats', 'Engine', and 'Mileage', while preserving 'Electric' cars:**

````python
# Filter the DataFrame to show only 'Electric' cars
electric_cars = df[df['Fuel_Type'] == 'Electric']

print("Details of 'Electric' cars before dropping rows with missing values of 'Seats', 'Engine', and 'Mileage':")
display(electric_cars[['Name', 'Fuel_Type', 'Seats', 'Engine', 'Mileage']])

# Check for missing values in the relevant columns for these electric cars
print("\nMissing values in 'Seats', 'Engine', 'Mileage' for 'Electric' cars:")
display(electric_cars[['Seats', 'Engine', 'Mileage']].isnull().sum())
````

````python
# Identify Electric cars
electric_car_indices = df[df['Fuel_Type'] == 'Electric'].index

# Impute 'Mileage' of Electric cars with 1.0 to preserve them from being dropped
df.loc[electric_car_indices, 'Mileage'] = df.loc[electric_car_indices, 'Mileage'].fillna(1.0)

# Drop rows with remaining missing values for Seats and Engine only
df.dropna(subset=['Seats', 'Engine'], inplace=True)

# Displaying info of updated data
df.info()
````

````python
# Checking number of missing values after this step
df.isnull().sum()
````

````python
# Checking the unique categories in each categorical variable again
for column in cat_cols:
    print(f"Column: {column} (Unique Categories: {df[column].nunique()})")
    print(df[column].value_counts())
    print("-" * 50)
````

**Step 2 - Imputing missing values of 'Power' using Brand from 'Name', 'Year', 'Fuel_Type','Transmission', 'Engine':**

**Substep - Extracting 'Brand' from the 'Name' column:**

````python
# Extracting 'Brand' from the 'Name' column
df['Brand'] = df['Name'].apply(lambda x: x.split(' ')[0])

# Checking the unique brands and their counts
print("Unique Brands and their Counts:")
print(df['Brand'].value_counts())

# Displaying the first few rows with the new 'Brand' column
print("\nDataFrame with 'Brand' column:")
display(df.head())
````

````python
# Getting the number of missing values of 'Power' before this imputation step
missing_power_before = df['Power'].isnull().sum()
print(f"Number of missing values of 'Power' before this imputation step: {missing_power_before}")

# Imputing missing values of 'Power' using 'Brand' from 'Name', 'Year', 'Fuel_Type','Transmission', 'Engine':**
df['Power'] = df.groupby(['Brand', 'Year', 'Fuel_Type', 'Transmission', 'Engine'])['Power'].transform(lambda x: x.fillna(x.median()))

# Checking the number of missing values after this step
missing_power_after = df['Power'].isnull().sum()
print(f"Number of missing values of 'Power' after this imputation step: {missing_power_after}")

# Calculating how many values were imputed in this step
imputed_values = missing_power_before - missing_power_after
print(f"Number of values of 'Power' imputed in this step: {imputed_values}")

# Calculating the percentage of imputed values relative to the initial number of missing values
percentage_imputed = (imputed_values / missing_power_before) * 100
print(f"Percentage of values imputed: {percentage_imputed:.2f}%")

# Displaying info of updated data
df.info()
````

**Step 3 - Dropping remaining 43 rows with missing values of 'Power':**

````python
# Dropping remaining 43 rows with missing values of Power
df.dropna(subset=['Power'], inplace=True)

# Displaying info of updated data
df.info()
````

````python
# Checking number of missing values after this step
df.isnull().sum()
````

**Step 4 - Impute 'New_price' using 'Brand', 'Year','Fuel_Type', 'Transmission', 'Engine', 'Power', and 'Seats' (in a phased approach):**

**Phase 1 - Impute 'New_price' using 'Brand', 'Year','Fuel_Type', 'Transmission', 'Engine', 'Power', and 'Seats**

````python
# Getting the number of missing values of 'New_price' before this imputation step
missing_New_price_before = df['New_price'].isnull().sum()
print(f"Number of missing values of 'New_price' before this imputation step: {missing_New_price_before}")

# Imputing missing values of New_price using Brand, Year, Fuel_Type, Transmission, Engine, Power, Seats:**
df['New_price'] = df.groupby(['Brand', 'Year', 'Fuel_Type', 'Transmission', 'Engine', 'Power', 'Seats'])['New_price'].transform(lambda x: x.fillna(x.median()))

# Checking the number of missing values after this step
missing_New_price_after = df['New_price'].isnull().sum()
print(f"Number of missing values of 'New_price' after this imputation step: {missing_New_price_after}")

# Calculating how many values were imputed in this step
imputed_values = missing_New_price_before - missing_New_price_after
print(f"Number of values of 'New_price' imputed in this step: {imputed_values}")

# Calculating the percentage of imputed values relative to the initial number of missing values
percentage_imputed = (imputed_values / missing_New_price_before) * 100
print(f"Percentage of values imputed: {percentage_imputed:.2f}%")

# Displaying info of updated data
df.info()
````

**Justification for broader 'New_price' Imputation (excluding 'Power'):**

* In the previous granular imputation for 'New_price', we included 'Power' as one of the grouping keys. While 'Power' is an important attribute, it is highly correlated with 'Engine' size. To a certain extent, a car's 'Power' output is largely determined by its 'Engine' capacity.

* When we use very granular groups (including 'Power'), we create very specific combinations of features. If a specific combination lacks sufficient existing 'New_price' values, then no median can be calculated, and the missing 'New_price' value cannot be imputed. This is why a significant number of 'New_price' values still remained missing after the first pass.

* By performing a second, broader imputation pass where we remove 'Power' from the grouping keys (but keep 'Brand', 'Year', 'Engine', 'Fuel_Type', 'Transmission', and 'Seats'), we expect to impute more of the remaining missing 'New_price' values. This is because:
  * **Reduces Granularity:** Removing 'Power' makes the grouping less specific, increasing the likelihood of finding non-missing 'New_price' values within each group to calculate a median.
  * **Leverages Correlation:** Since 'Engine' size largely captures the essence of 'Power' in determining a car's new price, we can still achieve reasonably accurate imputations without 'Power' in the grouping.
  * **Other Features Unaffected:** The other features (Brand, Year, Engine, Fuel_Type, Transmission, Seats) are largely independent of each other in a way that warrants their separate consideration for imputation groupings.

**Phase 2 - Broader Imputation for New_price without 'Power':**

````python
# Getting the number of missing values of 'New_price' before this imputation step
missing_New_price_withoutPower_before = df['New_price'].isnull().sum()
print(f"Number of missing values of 'New_price' before this imputation step: {missing_New_price_withoutPower_before}")

# Imputing missing values of New_price using Brand, Year, Fuel_Type, Transmission, Engine, Seats:**
df['New_price'] = df.groupby(['Brand', 'Year', 'Fuel_Type', 'Transmission', 'Engine', 'Seats'])['New_price'].transform(lambda x: x.fillna(x.median()))

# Checking the number of missing values of 'New_price' after this step
missing_New_price_withoutPower_after = df['New_price'].isnull().sum()
print(f"Number of missing values of 'New_price' after this imputation step: {missing_New_price_withoutPower_after}")

# Calculating how many values were imputed in this step
imputed_values = missing_New_price_withoutPower_before - missing_New_price_withoutPower_after
print(f"Number of values of 'New_price' imputed in this step: {imputed_values}")

# Calculating the percentage of imputed values relative to the initial number of missing values
percentage_imputed = (imputed_values / missing_New_price_withoutPower_before) * 100
print(f"Percentage of values imputed: {percentage_imputed:.2f}%")

# Displaying info of updated data
df.info()
````

**Function to plot a boxplot and histogram along the same scale:**

````python
# Function to plot a boxplot and histogram along the same scale

def histogram_boxplot(data, feature, figsize = (12, 7), kde = True, bins = None):
    """
    Boxplot and histogram combined

    data: dataframe
    feature: dataframe column
    figsize: size of figure (default (12,7))
    kde: whether to show density curve (default False)
    bins: number of bins for histogram (default None)
    """
    f2, (ax_box2, ax_hist2) = plt.subplots(
        nrows = 2,      # Number of rows of the subplot grid = 2
        sharex = True,  # x-axis will be shared among all subplots
        gridspec_kw = {"height_ratios": (0.25, 0.75)},
        figsize = figsize,
    )                   # Creating the 2 subplots
    sns.boxplot(
        data = data, x = feature, ax = ax_box2, showmeans = True, color = "violet"
    )                   # Boxplot will be created and a star will indicate the mean value of the column
    sns.histplot(
        data = data, x = feature, kde = kde, ax = ax_hist2, bins = bins, palette = "winter"
    ) if bins else sns.histplot(
        data = data, x = feature, kde = kde, ax = ax_hist2
    )                   # For histogram
    ax_hist2.axvline(
        data[feature].mean(), color = "green", linestyle = "--"
    )                   # Add mean to the histogram
    ax_hist2.axvline(
        data[feature].median(), color = "black", linestyle = "-"
    )                   # Add median to the histogram
````

**Log transformation of 'New_price' and Boxplot and Histogram of 'log_New_price':**

````python
# Log transformation of New_price:
df['log_New_price'] = np.log(df['New_price'])

# Plotting the boxplot and histogram for log_New_price to visualize the transformation
histogram_boxplot(df, 'log_New_price')
plt.show()
````

**Observations:**

* **Remaining Skewness:** While log transformation is powerful for reducing severe right-skewness, it doesn't always result in a perfectly normal distribution. Some residual skewness can persist if the original distribution was extremely skewed, or if there are other complex underlying patterns. The goal of transformation is often to make the distribution more symmetrical and manageable for models, rather than perfectly normal.
* **Two 'Peaks' (Bimodality):** Two peaks, or a 'bimodal' distribution, suggest that there might be two distinct subpopulations or clusters of new car prices within the dataset. Even after transformation, these two groups remain somewhat separated. These peaks may represent:
  * **Market Segments:** It's highly probable that these peaks correspond to different market segments. For instance, one peak might represent the distribution of new prices for economy/mid-range cars, while the other peak could represent the distribution of new prices for premium/luxury cars. These two segments naturally have different price points for new vehicles, leading to two distinct 'bumps' in the overall new price distribution.
  * **Vehicle Types:** Similarly, the peaks could correspond to broad categories of vehicles, like a cluster of smaller sedans/hatchbacks at one price point and a cluster of larger SUVs/MUVs at another, higher price point.

* **Bimodality Importance:** The bimodality is important information. It indicates that 'New_price' isn't just a single, continuous spectrum and likely contains natural divisions. For modeling, this means that the model might need to implicitly learn different relationships for these two distinct groups of cars. The log transformation helps to bring these two peaks closer and make their individual distributions more symmetric, but it doesn't erase the fact that they represent two different price 'centers'.
Bimodality in an independent variable doesn't strictly violate the core assumptions of linear regression in the same way, for example, non-linearity with the target variable or heteroscedasticity (unequal variance of errors) would. The primary assumptions are often about the errors being normally distributed, not necessarily the independent variables themselves.

* **Potential Impact on Model Performance:** However, bimodality can sometimes lead to:
  * **Reduced Model Fit:** A single linear relationship might not perfectly capture the dynamics if there are two distinct groups with different underlying relationships to the target variable. The model might struggle to fit both 'peaks' equally well.
  * **Residuals:** It might lead to non-normally distributed residuals or patterns in the residuals, which can affect the validity of statistical inferences (like p-values and confidence intervals).

**Options for Addressing Bimodality of 'log_New_price':**

1. **Keep as is (and evaluate performance):**
    * **Description:** Often, many machine learning models, especially tree-based ones, are robust enough to handle distributions that aren't perfectly normal. If the model can effectively learn the relationship within each mode, and the overall performance is good, explicit intervention might not be necessary.
    * **Rationale:** Simplicity is best, so starting with log_New_price directly and the evaluating the model's performance is the first option. If the model works well, further complexity might not be justified.

2. **Use Interaction Terms (if segments are identifiable):**
    * **Description:** If the two modes correspond to clearly identifiable segments (e.g., 'economy/mid-range' vs. 'premium/luxury' cars), then a new categorical feature can be created (e.g., is_premium_segment). Then, an interaction term between this new segment indicator and log_New_price can be introduced. This allows the model to learn different slopes and intercepts for log_New_price within each segment.
    * **Rationale:** This approach allows the model to capture different relationships between New_price and the target Price for different car segments, which is a powerful way to handle non-linearities that arise from distinct groups.

3. **Build Separate Models (for fundamentally different relationships):**
    * **Description:** In cases where the relationship between log_New_price and the used car Price is fundamentally different for each mode (e.g., depreciation curves for economy cars are vastly different from luxury cars), then building separate predictive models for each segment should be considered. This would require first classifing a car into its segment and then applying the corresponding model.
    * **Rationale:** This is the most complex approach but can yield the most accurate results if the underlying mechanisms governing pricing are truly distinct for each segment. It's often reserved for when other methods prove insufficient.

**Selected Approach:**

The chosen approach to addressing the bimodality of 'log_New_price' involves creating interaction terms based on identifiable segments. If the two modes in the 'log_New_price' distribution correspond to distinct segments, such as 'economy/mid-range' versus 'premium/luxury' cars, we can introduce a new categorical feature (e.g., is_premium_segment). Subsequently, an interaction term between this segment indicator and 'log_New_price' will be created.

The rationale behind this strategy is to enable the model to capture different relationships between the 'New_price' and the target 'Price' for these distinct car segments. This is a powerful method for handling non-linearities that arise from these separated groups within the data, leading to a more nuanced and accurate model.

**Creating Separate Categories for 'Mileage' by 'Fuel_Type':**

````python
# Creating a new grouped Fuel_Type column
def group_fuel_type(fuel):
    if fuel == 'Diesel':
        return 'Diesel'
    elif fuel == 'Petrol':
        return 'Petrol'
    elif fuel in ['CNG', 'LPG']:
        return 'Natural_Gas'
    elif fuel == 'Electric':
        return 'Electric'

df['Fuel_Type_Grouped'] = df['Fuel_Type'].apply(group_fuel_type)

# Display the value counts for the new column to verify
print(df['Fuel_Type_Grouped'].value_counts())
````

**Function to create labeled barplots:**

````python
# Function to create labeled barplots

def labeled_barplot(data, feature, perc = False, n = None):
    """
    Barplot with percentage at the top

    data: dataframe
    feature: dataframe column
    perc: whether to display percentages instead of count (default is False)
    n: displays the top n category levels (default is None, i.e., display all levels)
    """

    total = len(data[feature])            # Length of the column
    count = data[feature].nunique()
    if n is None:
        plt.figure(figsize = (count + 1, 5))
    else:
        plt.figure(figsize = (n + 1, 5))

    plt.xticks(rotation = 90, fontsize = 15)
    ax = sns.countplot(
        data = data,
        x = feature,
        palette = "Paired",
        order = data[feature].value_counts().index[:n].sort_values(),
    )

    for p in ax.patches:
        if perc == True:
            label = "{:.1f}%".format(
                100 * p.get_height() / total
            )                              # Percentage of each class of the category
        else:
            label = p.get_height()         # Count of each level of the category

        x = p.get_x() + p.get_width() / 2  # Width of the plot
        y = p.get_height()                 # Height of the plot

        ax.annotate(
            label,
            (x, y),
            ha = "center",
            va = "center",
            size = 12,
            xytext = (0, 5),
            textcoords = "offset points",
        )                                 # Annotate the percentage

    plt.show()                            # Show the plot
````

## **Feature Engineering**

**Verifying the current state of the DataFrame:**
* Handling of missing values, presence of log-transformed numerical features, and existence of engineered features. An overview of the DataFrame's structure, non-null counts, and data types.


````python
# Verifying the current state of the DataFrame
print("Current DataFrame Information:")
df.info()

print("\nVerifying specific engineered and transformed columns:")
required_columns = [
    'log_Price', 'log_Kilometers_Driven', 'log_Engine', 'log_Power', 'log_New_price',
    'Brand', 'Car_Age', 'Fuel_Type_Grouped'
]

for col in required_columns:
    if col in df.columns:
        print(f"- Column '{col}' exists and has dtype: {df[col].dtype}")
        if df[col].isnull().any():
            print(f"  (Warning: '{col}' still has missing values: {df[col].isnull().sum()})")
    else:
        print(f"- Column '{col}' DOES NOT EXIST.")
````

**Grouping `Owner_Type`:**

* Consolidating the 'Third' and 'Fourth & Above' categories within the 'Owner_Type' column into a new 'Multiple' category. This will create a revised 'Owner_Type_Revised' column for better handling of sparse categories.



````python
# Grouping 'Owner_Type'
def group_owner_type(owner):
    if owner in ['First', 'Second']:
        return owner
    else:
        return 'Multiple'

df['Owner_Type_Revised'] = df['Owner_Type'].apply(group_owner_type)

print("Value counts for new 'Owner_Type_Revised' column:")
print(df['Owner_Type_Revised'].value_counts())
````

#### **Exploring Advanced Feature Engineering (Interaction Terms, Non-linearity, and PCA)**

* Advanced feature engineering will involve generating squared and interaction terms from numerical variables. To manage multicollinearity, Principal Component Analysis (PCA) will be applied. Furthermore, a 'premium segment' indicator will be created for 'New_price' to address its bimodal nature.


**Creating a squared term for `Car_Age`:**
  * To capture potential non-linear relationships of age with price.



````python
# Creating a squared term for 'Car_Age'
df['Car_Age_Squared'] = df['Car_Age']**2

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Creating an interaction term between `Mileage` and `Car_Age`:**
  * To capture the combined effect of mileage and car age on price.



````python
# Creating an interaction term between 'Mileage' and 'Car_Age'
df['Mileage_x_Car_Age_Interaction'] = df['Mileage'] * df['Car_Age']

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Creating an interaction term between `log_Kilometers_Driven` and `Car_Age`:**

  * To capture the combined effect of kilometers driven and car age on price.

````python
# Creating an interaction term between 'log_Kilometers_Driven' and 'Car_Age'
df['log_Kilometers_Driven_x_Car_Age_Interaction'] = df['log_Kilometers_Driven'] * df['Car_Age']

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Creating a binary feature `Is_Premium_New_Price`:**
  * To capture the premium segment for new cars. This feature will be '1' if 'log_New_price' is above its median, and '0' otherwise, which addresses the bimodality observed earlier and categorizes cars based on their new price segment.

````python
# Creating a binary feature 'Is_Premium_New_Price'
median_log_new_price = df['log_New_price'].median()
df['Is_Premium_New_Price'] = (df['log_New_price'] > median_log_new_price).astype(int)

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Peforming Principal Components Analysis (PCA) for `Power` and `Engine`:**
  * To reduce multicollinearity between 'log_Power' and 'log_Engine' and capture their combined variance, (PCA) will be applied to these two features and add the first principal component will be a new feature named `Power_Engine_PCA`.  While PCA may successfully address the multicollinearity between log_Power and log_Engine, Power_Engine_PCA may remain correlated with log_New_price (as expected, since they are all strong indicators of a car's class/value). If VIF analysis suggests that the level of multicollinearity of Power_Engine_PCA within the overall model context was reduced to a more acceptable range, then no further action is needed. For linear models, regularization techniques like Ridge and Lasso are then particularly effective at managing any remaining multicollinearity by shrinking coefficients, even if it exists between Power_Engine_PCA and log_New_price.

````python
# Performing PCA for 'Power' and 'Engine'
from sklearn.decomposition import PCA

# Select the features for PCA
pca_features = ['log_Power', 'log_Engine']

# Handle potential NaN values before PCA (though they should be handled by previous steps)
# For PCA, it's crucial that there are no NaNs. If any still exist, a simple imputation (e.g., median) or dropping rows would be necessary.
# Based on previous info() outputs, log_Power and log_Engine have no missing values at this stage.

# Initialize PCA to reduce to 1 component
pca = PCA(n_components=1)

# Fit PCA and transform the data
df['Power_Engine_PCA'] = pca.fit_transform(df[pca_features])

# Display the head of the DataFrame with the new column
display(df.head())
````

**Creating an interaction term between `log_New_price` and `Care_Age`**:
  * To capture the combined effect of new price and car age on the used car price.



````python
# Creating an interaction term between 'log_New_price' and 'Care_Age'
df['log_New_price_x_Car_Age_Interaction'] = df['log_New_price'] * df['Car_Age']

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Creating an interaction term between `Kilometers_Driven` and `Car_Age`**:
  * To measure the average annual usage of a car. The goal is to capture the rate at which a car accumulates mileage over its lifespan. This rate itself is a meaningful feature, and expressing it in original units makes it a more direct and actionable insight.



````python
# Creating an interaction term between 'Kilometers_Driven' and 'Car_Age'
df['Kilometers_per_Year'] = df['Kilometers_Driven'] / df['Car_Age']

# Displaying the head of the DataFrame with the new column
display(df.head())
````

**Encoding Categorical Variables:**
  * Now that all numerical and interaction features have been created, the categorical variables will be encoded into a numerical format suitable for machine learning models and the original categorical columns along with the 'Name' column will be dropped.

````python
# Encoding Categorical Variables
import pandas as pd

# Identify categorical columns to be one-hot encoded
categorical_cols_to_encode = [
    'Location',
    'Fuel_Type_Grouped',
    'Transmission',
    'Owner_Type_Revised',
    'Brand',
    'Seats' # Treat 'Seats' as categorical for encoding due to non-linear relationship
]

# Apply one-hot encoding
df = pd.get_dummies(df, columns=categorical_cols_to_encode, drop_first=True)

# Drop the original 'Name' column
df.drop('Name', axis=1, inplace=True)

# Display the head of the DataFrame with the new columns and dropped 'Name'
display(df.head())

# Display the info of the DataFrame to verify the new columns and data types
df.info()
````

**Dropping `Fuel_Type` and `Owner_Type`**:
  * The 'Fuel_Type' and 'Owner_Type' columns are redundant after their grouped/revised versions were created and one-hot encoded. These original columns will be dropped to finalize the dataset for modeling, ensuring all features are in a numerical format suitable for machine learning algorithms.


````python
# Dropping 'Fuel_Type' and 'Owner_Type'
df.drop('Fuel_Type', axis=1, inplace=True)
df.drop('Owner_Type', axis=1, inplace=True)

# Display the head of the DataFrame with the new columns and dropped 'Name'
display(df.head())

# Display the info of the DataFrame to verify the new columns and data types
df.info()
````

# **Important Insights from EDA and Data Preprocessing**

What are the most important observations and insights from the data based on the EDA and Data Preprocessing performed?

## **Important Insights from EDA and Data Preprocessing**

**Based on the Exploratory Data Analysis (EDA) and the subsequent data preprocessing steps, several critical observations and insights have emerged that will guide the modeling phase:**

### **1. Data Integrity and Cleaning:**

*   **Missing Target Variable (`Price`):** Rows with missing 'Price' were dropped, reducing the dataset size from 7253 to 6019 observations, ensuring a reliable target for supervised learning.
*   **Handling Key Feature Missing Values:**
    *   `Mileage` for 'Electric' cars (2 observations) was imputed to '1.0' (which becomes '0' after log transformation) to retain these vehicles, acknowledging their unique characteristics and growing market segment.
    *   Rows with missing `Seats` (42) and `Engine` (36) were dropped, as these are critical structural attributes difficult to impute reliably, resulting in 5934 observations.
    *   `Power` missing values (103) were largely imputed (60 values) using group medians ('Brand', 'Year', 'Fuel_Type_Grouped', 'Transmission', 'Engine'), and the remaining 43 rows were dropped, leading to 5897 observations.
    *   An extreme outlier in `Kilometers_Driven` (6.5 million km) was identified and removed.
    *   36 rows with '0.0' `Mileage` (anomalous for fuel-burning cars) were identified and dropped, refining the dataset to 5897 observations.
*   **High Missingness in `New_price`:** Despite a two-phase imputation strategy, 'New_price' still has significant missing values (~55%). This necessitates building models both with and without 'New_price' to assess its trade-off between predictive power and data quantity.

### **2. Feature Engineering and Transformation:**

*   **Log Transformations for Skewed Numerical Features:**
    *   'Price' (target), 'New_price', 'Kilometers_Driven', 'Engine', and 'Power' were all highly right-skewed. Log transformation (into 'log_Price', 'log_New_price', 'log_Kilometers_Driven', 'log_Engine', 'log_Power' respectively) effectively normalized their distributions, making them more suitable for linear models.
    *   The 'bumpiness' in 'log_Engine' was identified as a reflection of common, standardized engine capacities in the market, not a flaw.
*   **Derived Numerical Features:**
    *   `Car_Age` (Current Year - 'Year') was created, providing a more intuitive measure of a car's age than the manufacturing year.
    *   `Car_Age_Squared` was created to capture non-linear depreciation patterns.
    *   `Kilometers_per_Year` was created (`Kilometers_Driven` / `Car_Age`) to represent average annual usage.
    *   Interaction terms like `Mileage_x_Car_Age_Interaction`, `log_Kilometers_Driven_x_Car_Age_Interaction`, and `log_New_price_x_Car_Age_Interaction` were created to capture combined effects of features.
*   **Categorical Feature Refinements:**
    *   `Brand` was extracted from 'Name', reducing the cardinality of a highly unique identifier while retaining valuable brand information.
    *   `Fuel_Type_Grouped` was created, combining 'CNG' and 'LPG' into 'Natural_Gas' while keeping 'Petrol', 'Diesel', and 'Electric' distinct, reflecting a practical grouping strategy.
    *   `Owner_Type_Revised` was created, combining 'Third' and 'Fourth & Above' into 'Multiple' due to sparsity and to create a more robust 'well-used' category.
    *   `Seats` was identified as a discrete variable with a potentially non-linear relationship to price; it was treated as a categorical variable for encoding.
*   **Dimensionality Reduction:** PCA was applied to 'log_Power' and 'log_Engine' to create `Power_Engine_PCA`, addressing high multicollinearity between these two features.
*   **One-Hot Encoding:** All relevant categorical features ('Location', 'Fuel_Type_Grouped', 'Transmission', 'Owner_Type_Revised', 'Brand', 'Seats') were one-hot encoded.

### **3. Key Relationships and Predictors ('log_Price'):**

*   **Strong Positive Predictors:** `log_New_price` (**very strong, 0.87**), `log_Power` (**strong, 0.79**), and `log_Engine` (**strong, 0.69**) are the most influential numerical features, indicating that new price, power, and engine size are primary drivers of used car prices.
*   **Moderate Negative Predictors:** `Car_Age` (-0.49) and `Mileage` (-0.31) show expected depreciation effects.
*   **Weak Predictors:** `log_Kilometers_Driven` (-0.20) and `Seats` (0.17) have weaker linear correlations but are still important to consider.
*   **Multicollinearity:** Significant multicollinearity exists among `log_New_price`, `log_Power`, and `log_Engine`, which was addressed by PCA to manage their combined effect in linear models.
*   **Categorical Influence:** All categorical variables ('Location', 'Fuel_Type_Grouped', 'Transmission', 'Owner_Type_Revised', 'Brand') demonstrate a clear impact on 'log_Price' and will be valuable features after encoding.
*   **Outlier Identification:** An 'Electric' car ('Mahindra E Verito D4') was identified as an outlier in the 'log_Price' vs. 'log_Engine' plot, which is a legitimate data point reflecting its unique engine characteristics rather than an error.

## **Modeling Strategy**

To effectively evaluate the impact of various preprocessing choices and feature engineering techniques on the predictive performance of the models, a comprehensive modeling strategy has been devised. This strategy involves creating multiple DataFrames, each with a distinct set of features, allowing for a systematic comparison.

The rationale behind this approach is as follows:

1.  **Log-transformed vs. Original Numerical Features**: Many numerical features (`Price`, `New_price`, `Kilometers_Driven`, `Engine`, `Power`) were found to be highly skewed. Log-transformation helps to normalize their distributions, stabilize variance, and mitigate the influence of outliers, which can significantly improve the performance and assumptions of linear models. By comparing models built with both original and log-transformed features, the benefits of this transformation can be quantified.

2.  **Inclusion/Exclusion of `New_price`**: The 'New_price' feature (and its log-transformed version, `log_New_price`) is theoretically a very strong predictor of used car prices. However, it suffers from significant missing values (~55%), even after extensive imputation. To understand the trade-off between the predictive power of this feature and the reduction in dataset size (or reliance on imputed values), models will be built both with and without `log_New_price`. This will help determine if the gains from using this feature outweigh the challenges posed by its missingness.

3.  **Incorporation of Advanced Engineered Features**: To capture more complex relationships and potentially improve model accuracy, advanced features have been engineered. These include:
    *   **Squared terms**: `Car_Age_Squared` to model non-linear depreciation.
    *   **Interaction terms**: `Mileage_x_Car_Age_Interaction`, `log_Kilometers_Driven_x_Car_Age_Interaction`, and `log_New_price_x_Car_Age_Interaction` to capture combined effects.
    *   **Ratio features**: `Kilometers_per_Year` to represent annual usage.
    *   **Categorical treatment of 'Seats'**: Recognizing the non-linear relationship of seats with price, 'Seats' has been one-hot encoded as a categorical variable.
    * By comparing models with and without these advanced features, their incremental value can be assessed.

4.  **Application vs. Non-Application of PCA**: `log_Power`, `log_Engine`, and `log_New_price` were found to be highly inter-correlated, indicating multicollinearity. Principal Component Analysis (PCA) was applied to `log_Power` and `log_Engine` to create `Power_Engine_PCA`, aiming to reduce multicollinearity and capture their combined variance efficiently. Models will be tested with `Power_Engine_PCA` in place of individual `log_Power` and `log_Engine` to see if it leads to more stable models or improved performance, especially for linear regression techniques.

This multi-faceted approach will allow for a robust evaluation, providing insights into which feature sets yield the most accurate and interpretable models for used car price prediction.

## **Implementing OLS Model Evaluation Function with Metrics, CV, and Bootstrapping**




#### **Creating Function for OLS Models:**

  * Developing a robust Python function that, for a given DataFrame and target variable, performs OLS regression modeling, evaluates performance metrics (R-squared, Adjusted R-squared, MSE, RMSE, MAE, MAPE) with K-fold Cross-Validation and Bootstrapping for confidence intervals, conducts OLS diagnostics (p-values, heteroskedasticity tests, VIF), and visualizes results.




````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_ols_model(
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95
):
    """
    Performs OLS regression modeling, evaluates performance metrics, conducts OLS diagnostics,
    and visualizes results with K-fold Cross-Validation and Bootstrapping for confidence intervals.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results (e.g., 0.95 for 95% CI).

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              diagnostic results, and generated plots.
    """

    results = {}

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for statsmodels
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Add a constant to the features for OLS model
    X = sm.add_constant(X)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Fit an OLS model using statsmodels.api.OLS on the training data
    model = sm.OLS(y_train, X_train)
    model_results = model.fit()

    # Obtain model predictions
    y_train_pred = model_results.predict(X_train)
    y_test_pred = model_results.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred, n_features):
        r2 = r2_score(y_true, y_pred)
        n = len(y_true)
        p = n_features - 1  # -1 because of the added constant
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        # MAPE calculation (handle division by zero if actual values are 0)
        mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
        # Replace inf with nan if y_true contains zeros
        mape = np.nan_to_num(mape, nan=np.inf) if np.isinf(mape).any() else mape

        return r2, adj_r2, mse, rmse, mae, mape

    # 4. Calculate and print metrics for training and testing sets
    n_features_train = X_train.shape[1]
    r2_train, adj_r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred, n_features_train)
    r2_test, adj_r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred, X_test.shape[1])

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train Adjusted R-squared: {adj_r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test Adjusted R-squared: {adj_r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'Adj_R2': adj_r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'Adj_R2': adj_r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 5. K-fold Cross-Validation
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_adj_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]

        model_fold = sm.OLS(y_train_fold, X_train_fold).fit()
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, adj_r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold, X_val_fold.shape[1])
        cv_r2.append(r2_fold)
        cv_adj_r2.append(adj_r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'Adj_R2_mean': np.mean(cv_adj_r2), 'Adj_R2_std': np.std(cv_adj_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))


    # 6. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae = [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train), len(X_train), replace=True)
        X_boot, y_boot = X_train.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit OLS model on bootstrap sample
        try:
            model_boot = sm.OLS(y_boot, X_boot).fit()
        except Exception as e:
            # Handle cases where bootstrap sample might be degenerate
            # print(f"Warning: OLS failed for bootstrap sample {i}. Skipping. Error: {e}")
            continue

        # Predict on original test set
        y_test_pred_boot = model_boot.predict(X_test)

        # Calculate and store metrics
        r2_boot, _, mse_boot, rmse_boot, mae_boot, _ = calculate_metrics(y_test, y_test_pred_boot, X_test.shape[1])
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)

        # Store coefficients
        for col in X_train.columns:
            if col in model_boot.params:
                boot_coefficients[col].append(model_boot.params[col])
            else:
                boot_coefficients[col].append(np.nan) # Append nan if coefficient not found (e.g., due to perfect multicollinearity in bootstrap sample)


    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE'], [boot_r2, boot_mse, boot_rmse, boot_mae]):
        # Filter out NaN values before calculating percentiles
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    for col, ci_vals in coefficients_ci.items():
        print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")


    # 7. OLS Diagnostics
    print("\n--- OLS Model Summary (Full Training Data) ---")
    print(model_results.summary())
    results['ols_summary'] = model_results.summary().as_html()

    # Breusch-Pagan test for heteroskedasticity
    print("\n--- Breusch-Pagan Test for Heteroskedasticity ---")
    # The Breusch-Pagan test requires model residuals and the exogenous variables (X)
    # The test needs the original X_train, not X_train with constant for formula, but sm.add_constant() is standard.
    # Using the residual of the model and original X_train for the test
    names = ['Lagrange multiplier statistic', 'p-value', 'f-value', 'f p-value']
    bp_test = het_breuschpagan(model_results.resid, model_results.model.exog)
    print(pd.Series(bp_test, index=names))
    results['breusch_pagan_test'] = pd.Series(bp_test, index=names).to_dict()

    # Variance Inflation Factor (VIF)
    print("\n--- Variance Inflation Factor (VIF) ---")
    vif_data = pd.DataFrame()
    vif_data["feature"] = X_train.columns
    vif_data["VIF"] = [variance_inflation_factor(X_train.values, i) for i in range(X_train.shape[1])]
    print(vif_data.sort_values(by="VIF", ascending=False))
    results['vif'] = vif_data.sort_values(by="VIF", ascending=False).to_dict('records')

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title('Test Set: Actual vs. Predicted Prices')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title('Test Set: Residuals vs. Predicted Prices')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']), reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')

            plt.tight_layout()
            plt.show()

    return results
````

#### **Updating the function to handle `inf` and `nan` values more robustly.**




````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_ols_model(
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95
):
    """
    Performs OLS regression modeling, evaluates performance metrics, conducts OLS diagnostics,
    and visualizes results with K-fold Cross-Validation and Bootstrapping for confidence intervals.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results (e.g., 0.95 for 95% CI).

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              diagnostic results, and generated plots.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for statsmodels
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Add a constant to the features for OLS model
    X = sm.add_constant(X)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Fit an OLS model using statsmodels.api.OLS on the training data
    model = sm.OLS(y_train, X_train)
    model_results = model.fit()

    # Obtain model predictions
    y_train_pred = model_results.predict(X_train)
    y_test_pred = model_results.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred, n_features):
        r2 = r2_score(y_true, y_pred)
        n = len(y_true)
        p = n_features - 1  # -1 because of the added constant
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        # MAPE calculation (handle division by zero if actual values are 0)
        # Ensure y_true does not contain zeros for MAPE calculation, or handle them gracefully.
        # For log-transformed prices, y_true should not be zero, but we add robustness.
        mape_values = np.abs((y_true - y_pred) / y_true)
        # Replace inf with nan if y_true contains zeros that lead to inf
        mape_values = np.where(y_true == 0, np.nan, mape_values)
        mape = np.nanmean(mape_values) * 100 # Use nanmean to ignore NaNs resulting from y_true=0

        return r2, adj_r2, mse, rmse, mae, mape

    # 4. Calculate and print metrics for training and testing sets
    n_features_train = X_train.shape[1]
    r2_train, adj_r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred, n_features_train)
    r2_test, adj_r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred, X_test.shape[1])

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train Adjusted R-squared: {adj_r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test Adjusted R-squared: {adj_r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'Adj_R2': adj_r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'Adj_R2': adj_r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 5. K-fold Cross-Validation
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_adj_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]

        model_fold = sm.OLS(y_train_fold, X_train_fold).fit()
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, adj_r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold, X_val_fold.shape[1])
        cv_r2.append(r2_fold)
        cv_adj_r2.append(adj_r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'Adj_R2_mean': np.mean(cv_adj_r2), 'Adj_R2_std': np.std(cv_adj_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))


    # 6. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae = [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train), len(X_train), replace=True)
        X_boot, y_boot = X_train.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit OLS model on bootstrap sample
        try:
            model_boot = sm.OLS(y_boot, X_boot).fit()
        except Exception as e:
            # Handle cases where bootstrap sample might be degenerate
            # print(f"Warning: OLS failed for bootstrap sample {i}. Skipping. Error: {e}")
            continue

        # Predict on original test set
        y_test_pred_boot = model_boot.predict(X_test)

        # Calculate and store metrics
        r2_boot, _, mse_boot, rmse_boot, mae_boot, _ = calculate_metrics(y_test, y_test_pred_boot, X_test.shape[1])
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)

        # Store coefficients
        for col in X_train.columns:
            if col in model_boot.params:
                boot_coefficients[col].append(model_boot.params[col])
            else:
                boot_coefficients[col].append(np.nan) # Append nan if coefficient not found (e.g., due to perfect multicollinearity in bootstrap sample)


    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE'], [boot_r2, boot_mse, boot_rmse, boot_mae]):
        # Filter out NaN values before calculating percentiles
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    for col, ci_vals in coefficients_ci.items():
        print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")


    # 7. OLS Diagnostics
    print("\n--- OLS Model Summary (Full Training Data) ---")
    print(model_results.summary())
    results['ols_summary'] = model_results.summary().as_html()

    # Breusch-Pagan test for heteroskedasticity
    print("\n--- Breusch-Pagan Test for Heteroskedasticity ---")
    # The Breusch-Pagan test requires model residuals and the exogenous variables (X)
    # The test needs the original X_train, not X_train with constant for formula, but sm.add_constant() is standard.
    # Using the residual of the model and original X_train for the test
    names = ['Lagrange multiplier statistic', 'p-value', 'f-value', 'f p-value']
    bp_test = het_breuschpagan(model_results.resid, model_results.model.exog)
    print(pd.Series(bp_test, index=names))
    results['breusch_pagan_test'] = pd.Series(bp_test, index=names).to_dict()

    # Variance Inflation Factor (VIF)
    print("\n--- Variance Inflation Factor (VIF) ---")
    vif_data = pd.DataFrame()
    vif_data["feature"] = X_train.columns
    vif_data["VIF"] = [variance_inflation_factor(X_train.values, i) for i in range(X_train.shape[1])]
    print(vif_data.sort_values(by="VIF", ascending=False))
    results['vif'] = vif_data.sort_values(by="VIF", ascending=False).to_dict('records')

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title('Test Set: Actual vs. Predicted Prices')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title('Test Set: Residuals vs. Predicted Prices')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']), reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')

            plt.tight_layout()
            plt.show()

    return results
````

These methods are designed to combat multicollinearity and prevent overfitting, leading to more stable and generalizable models.

### **1. Importance of Addressing Heteroskedasticity in OLS Models:**

**Heteroskedasticity** refers to the situation where the variance of the residuals (errors) in a regression model is not constant across all levels of the independent variables. The OLS diagnostics (Breusch-Pagan test) consistently indicated its significant presence in all models, including the best OLS model. While heteroskedasticity does not bias the coefficient estimates themselves, it does lead to:

*   **Biased Standard Errors**: The standard errors of the coefficients become incorrect, typically underestimated. This means the t-statistics and p-values are unreliable.
*   **Incorrect Statistical Inferences**: Due to biased standard errors, one might incorrectly conclude that some predictors are statistically significant when they are not, or fail to recognize significance when it actually exists. This impacts the reliability of hypothesis testing and confidence intervals for coefficients.
*   **Inefficient Estimates**: The OLS estimates are no longer the most efficient (i.e., they don't have the smallest variance among unbiased estimators).

**How Robust Standard Errors Help**: By using **robust standard errors** (also known as Huber-White standard errors), the estimated standard errors can be corrected for heteroskedasticity. This allows for:

*   **Obtain Reliable p-values and Confidence Intervals**: With corrected standard errors, the t-statistics and p-values become more accurate, enabling reliable statistical inferences about the significance of each predictor.
*   **Preserve Coefficient Estimates**: Robust standard errors do not change the estimated regression coefficients themselves. This means the model's interpretation in terms of the magnitude and direction of feature impacts remains the same, but confidence in these interpretations is improved.

Therefore, applying robust standard errors to the best OLS model is crucial for drawing statistically sound conclusions about the relationships between the features and the `log_Price`.

### **2. Introduction of Ridge and Lasso Regularization:**

**Ridge and Lasso regression** are regularization techniques that extend linear regression by adding a penalty term to the loss function during training. These techniques are introduced to address two primary issues identified in the OLS models: **multicollinearity** and **overfitting**.

*   **Handling Multicollinearity**: The OLS diagnostics (VIF analysis) indicated severe multicollinearity, especially among the one-hot encoded 'Brand' and 'Seats' features, and noticeable correlation between `log_Power` and `log_Engine` (though reduced by PCA). Multicollinearity can lead to unstable and highly sensitive coefficient estimates in OLS, making it difficult to interpret the individual impact of correlated predictors.
    *   **Ridge Regression (L2 regularization)**: Adds a penalty proportional to the square of the magnitude of coefficients. It shrinks coefficients towards zero, but rarely to exactly zero. This effectively reduces the impact of multicollinearity by spreading the influence among correlated predictors, leading to more stable estimates.
    *   **Lasso Regression (L1 regularization)**: Adds a penalty proportional to the absolute value of the magnitude of coefficients. Unlike Ridge, Lasso can shrink some coefficients exactly to zero, effectively performing feature selection. This is particularly useful when many features might be correlated or redundant, helping to build sparser, more interpretable models.

*   **Preventing Overfitting**: While the OLS models showed good R-squared values on both train and test sets, the complexity added by numerous features (especially after one-hot encoding many categorical variables) still poses a risk of overfitting to the training data. Regularization techniques impose a constraint on the magnitude of the coefficients, discouraging overly complex models that might perform poorly on unseen data.

By employing Ridge and Lasso, more robust models can be built that are less sensitive to correlated predictors and have better generalization performance on new, unseen used car data.

### **3. Selection of DataFrames for Regularization:**

Ridge and Lasso regularization will be applied to the following two DataFrames:

1.  **`df_transformed_with_new_price_extended_with_pca`**: This DataFrame incorporates all the best preprocessing and feature engineering efforts, including log-transformed numerical features, advanced engineered terms, PCA for `log_Power` and `log_Engine`, and crucially, the `log_New_price` feature. It was identified as the best-performing OLS model, explaining the most variance in `log_Price`.
2.  **`df_transformed_no_new_price_extended_with_pca`**: This DataFrame represents the best feature set *without* the `log_New_price` feature. It includes log-transformed numerical features, advanced engineered terms, and PCA for `log_Power` and `log_Engine`. It was noted that the `log_New_price` feature significantly reduces the number of available samples due to its high missingness. By regularizing this DataFrame, it can be assessed if a model trained on a larger dataset (without `log_New_price`) but with regularization can achieve comparable or better generalized performance, especially if the impact of `log_New_price` is overemphasized due to its presence in the training set.

This comparative strategy will allow for assessing the trade-offs between model complexity, data size, and the benefits of regularization, especially regarding the influential `log_New_price` feature.

### **4. Comparison of Regularized Models with Best OLS Model:**

The performance of the Ridge and Lasso models will be rigorously compared against the best OLS model (`df_transformed_with_new_price_extended_with_pca`, now with robust standard errors). The comparison will focus on:

*   **Accuracy Metrics**: R-squared, Adjusted R-squared, RMSE, MAE, and MAPE on the test set. Regularization is expected to potentially improve these metrics, particularly RMSE and MAE, due to better generalization.
*   **Stability**: Evaluation of cross-validation and bootstrapping results for all metrics to assess the consistency and reliability of predictions across different data subsets.
*   **Coefficient Magnitudes and Sparsity**: For Lasso, it will be observed which coefficients are driven to zero, indicating feature importance and potential for model simplification. For Ridge, the shrinkage of coefficients will be examined.
*   **Robustness to Multicollinearity**: The VIFs will be examined for the stability of coefficients in the presence of correlated predictors to confirm if regularization has mitigated these issues more effectively than OLS.
*   **Interpretability**: The impact of regularization on coefficient interpretability will be analyzed, especially for Lasso's feature selection capability.

This comprehensive comparison will ultimately help determine the most accurate, stable, and robust predictive model for used car prices, considering both statistical rigor and practical applicability.

## **Addressing Heteroskedasticity in Best OLS Model with Robust Standard Errors**

The best-performing OLS model will be re-evaluated (`df_transformed_with_new_price_extended_with_pca`) using robust standard errors (`cov_type='HC3'`) within `statsmodels.OLS` to address the identified heteroskedasticity. This will provide corrected p-values and confidence intervals for the coefficients, improving statistical inference without changing the coefficient estimates.


#### **Modifing the `evaluate_ols_model` function** for specifying robust standard errors.



````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_ols_model(
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95,
    cov_type='nonrobust' # Added cov_type parameter with default
):
    """
    Performs OLS regression modeling, evaluates performance metrics, conducts OLS diagnostics,
    and visualizes results with K-fold Cross-Validation and Bootstrapping for confidence intervals.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results (e.g., 0.95 for 95% CI).
        cov_type (str): Type of covariance matrix estimator to use (e.g., 'nonrobust', 'HC0', 'HC1', 'HC3').

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              diagnostic results, and generated plots.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for statsmodels
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Add a constant to the features for OLS model
    X = sm.add_constant(X)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Fit an OLS model using statsmodels.api.OLS on the training data
    # Pass cov_type to the fit method
    model = sm.OLS(y_train, X_train)
    model_results = model.fit(cov_type=cov_type)

    # Obtain model predictions
    y_train_pred = model_results.predict(X_train)
    y_test_pred = model_results.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred, n_features):
        r2 = r2_score(y_true, y_pred)
        n = len(y_true)
        p = n_features - 1  # -1 because of the added constant
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        # MAPE calculation (handle division by zero if actual values are 0)
        # Ensure y_true does not contain zeros for MAPE calculation, or handle them gracefully.
        # For log-transformed prices, y_true should not be zero, but we add robustness.
        mape_values = np.abs((y_true - y_pred) / y_true)
        # Replace inf with nan if y_true contains zeros that lead to inf
        mape_values = np.where(y_true == 0, np.nan, mape_values)
        mape = np.nanmean(mape_values) * 100 # Use nanmean to ignore NaNs resulting from y_true=0

        return r2, adj_r2, mse, rmse, mae, mape

    # 4. Calculate and print metrics for training and testing sets
    n_features_train = X_train.shape[1]
    r2_train, adj_r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred, n_features_train)
    r2_test, adj_r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred, X_test.shape[1])

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train Adjusted R-squared: {adj_r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test Adjusted R-squared: {adj_r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'Adj_R2': adj_r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'Adj_R2': adj_r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 5. K-fold Cross-Validation
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_adj_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]

        # Pass cov_type to the fit method within CV loop
        model_fold = sm.OLS(y_train_fold, X_train_fold).fit(cov_type=cov_type)
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, adj_r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold, X_val_fold.shape[1])
        cv_r2.append(r2_fold)
        cv_adj_r2.append(adj_r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'Adj_R2_mean': np.mean(cv_adj_r2), 'Adj_R2_std': np.std(cv_adj_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))


    # 6. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae = [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train), len(X_train), replace=True)
        X_boot, y_boot = X_train.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit OLS model on bootstrap sample
        try:
            # Pass cov_type to the fit method within bootstrap loop
            model_boot = sm.OLS(y_boot, X_boot).fit(cov_type=cov_type)
        except Exception as e:
            # Handle cases where bootstrap sample might be degenerate
            # print(f"Warning: OLS failed for bootstrap sample {i}. Skipping. Error: {e}")
            continue

        # Predict on original test set
        y_test_pred_boot = model_boot.predict(X_test)

        # Calculate and store metrics
        r2_boot, _, mse_boot, rmse_boot, mae_boot, _ = calculate_metrics(y_test, y_test_pred_boot, X_test.shape[1])
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)

        # Store coefficients
        for col in X_train.columns:
            if col in model_boot.params:
                boot_coefficients[col].append(model_boot.params[col])
            else:
                boot_coefficients[col].append(np.nan) # Append nan if coefficient not found (e.g., due to perfect multicollinearity in bootstrap sample)


    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE'], [boot_r2, boot_mse, boot_rmse, boot_mae]):
        # Filter out NaN values before calculating percentiles
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    for col, ci_vals in coefficients_ci.items():
        print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")


    # 7. OLS Diagnostics
    print("\n--- OLS Model Summary (Full Training Data) ---")
    print(model_results.summary()) # This summary will now use the specified cov_type
    results['ols_summary'] = model_results.summary().as_html()

    # Breusch-Pagan test for heteroskedasticity
    print("\n--- Breusch-Pagan Test for Heteroskedasticity ---")
    # The Breusch-Pagan test requires model residuals and the exogenous variables (X)
    # The test needs the original X_train, not X_train with constant for formula, but sm.add_constant() is standard.
    # Using the residual of the model and original X_train for the test
    names = ['Lagrange multiplier statistic', 'p-value', 'f-value', 'f p-value']
    bp_test = het_breuschpagan(model_results.resid, model_results.model.exog)
    print(pd.Series(bp_test, index=names))
    results['breusch_pagan_test'] = pd.Series(bp_test, index=names).to_dict()

    # Variance Inflation Factor (VIF)
    print("\n--- Variance Inflation Factor (VIF) ---")
    vif_data = pd.DataFrame()
    vif_data["feature"] = X_train.columns
    # Drop 'const' from VIF calculation as it causes inf and is not a true predictor
    X_train_no_const = X_train.drop(columns=['const'])
    if not X_train_no_const.empty:
        vif_data["VIF"] = [variance_inflation_factor(X_train_no_const.values, i) for i in range(X_train_no_const.shape[1])]
        # Map VIFs back to original columns, handling 'const' separately
        vif_data = pd.DataFrame({'feature': X_train_no_const.columns, 'VIF': [variance_inflation_factor(X_train_no_const.values, i) for i in range(X_train_no_const.shape[1])]})
        vif_data = pd.concat([pd.DataFrame([{'feature': 'const', 'VIF': np.nan}]), vif_data]).reset_index(drop=True)
    else:
        vif_data["VIF"] = np.nan # No features to calculate VIF for

    print(vif_data.sort_values(by="VIF", ascending=False))
    results['vif'] = vif_data.sort_values(by="VIF", ascending=False).to_dict('records')

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title('Test Set: Actual vs. Predicted Prices')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title('Test Set: Residuals vs. Predicted Prices')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']), reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')

            plt.tight_layout()
            plt.show()

    return results
````

#### **Updating the`evaluate_ols_model` function with a corrected VIF calculation.**


````python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_ols_model(
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95,
    cov_type='nonrobust' # Added cov_type parameter with default
):
    """
    Performs OLS regression modeling, evaluates performance metrics, conducts OLS diagnostics,
    and visualizes results with K-fold Cross-Validation and Bootstrapping for confidence intervals.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results (e.g., 0.95 for 95% CI).
        cov_type (str): Type of covariance matrix estimator to use (e.g., 'nonrobust', 'HC0', 'HC1', 'HC3').

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              diagnostic results, and generated plots.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for statsmodels
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Add a constant to the features for OLS model
    X = sm.add_constant(X)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Fit an OLS model using statsmodels.api.OLS on the training data
    # Pass cov_type to the fit method
    model = sm.OLS(y_train, X_train)
    model_results = model.fit(cov_type=cov_type)

    # Obtain model predictions
    y_train_pred = model_results.predict(X_train)
    y_test_pred = model_results.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred, n_features):
        r2 = r2_score(y_true, y_pred)
        n = len(y_true)
        p = n_features - 1  # -1 because of the added constant
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        # MAPE calculation (handle division by zero if actual values are 0)
        # Ensure y_true does not contain zeros for MAPE calculation, or handle them gracefully.
        # For log-transformed prices, y_true should not be zero, but we add robustness.
        mape_values = np.abs((y_true - y_pred) / y_true)
        # Replace inf with nan if y_true contains zeros that lead to inf
        mape_values = np.where(y_true == 0, np.nan, mape_values)
        mape = np.nanmean(mape_values) * 100 # Use nanmean to ignore NaNs resulting from y_true=0

        return r2, adj_r2, mse, rmse, mae, mape

    # 4. Calculate and print metrics for training and testing sets
    n_features_train = X_train.shape[1]
    r2_train, adj_r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred, n_features_train)
    r2_test, adj_r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred, X_test.shape[1])

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train Adjusted R-squared: {adj_r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test Adjusted R-squared: {adj_r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'Adj_R2': adj_r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'Adj_R2': adj_r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 5. K-fold Cross-Validation
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_adj_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]

        # Pass cov_type to the fit method within CV loop
        model_fold = sm.OLS(y_train_fold, X_train_fold).fit(cov_type=cov_type)
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, adj_r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold, X_val_fold.shape[1])
        cv_r2.append(r2_fold)
        cv_adj_r2.append(adj_r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'Adj_R2_mean': np.mean(cv_adj_r2), 'Adj_R2_std': np.std(cv_adj_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))


    # 6. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae = [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train), len(X_train), replace=True)
        X_boot, y_boot = X_train.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit OLS model on bootstrap sample
        try:
            # Pass cov_type to the fit method within bootstrap loop
            model_boot = sm.OLS(y_boot, X_boot).fit(cov_type=cov_type)
        except Exception as e:
            # Handle cases where bootstrap sample might be degenerate
            # print(f"Warning: OLS failed for bootstrap sample {i}. Skipping. Error: {e}")
            continue

        # Predict on original test set
        y_test_pred_boot = model_boot.predict(X_test)

        # Calculate and store metrics
        r2_boot, _, mse_boot, rmse_boot, mae_boot, _ = calculate_metrics(y_test, y_test_pred_boot, X_test.shape[1])
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)

        # Store coefficients
        for col in X_train.columns:
            if col in model_boot.params:
                boot_coefficients[col].append(model_boot.params[col])
            else:
                boot_coefficients[col].append(np.nan) # Append nan if coefficient not found (e.g., due to perfect multicollinearity in bootstrap sample)


    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE'], [boot_r2, boot_mse, boot_rmse, boot_mae]):
        # Filter out NaN values before calculating percentiles
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    for col, ci_vals in coefficients_ci.items():
        print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")


    # 7. OLS Diagnostics
    print("\n--- OLS Model Summary (Full Training Data) ---")
    print(model_results.summary()) # This summary will now use the specified cov_type
    results['ols_summary'] = model_results.summary().as_html()

    # Breusch-Pagan test for heteroskedasticity
    print("\n--- Breusch-Pagan Test for Heteroskedasticity ---")
    # The Breusch-Pagan test requires model residuals and the exogenous variables (X)
    # The test needs the original X_train, not X_train with constant for formula, but sm.add_constant() is standard.
    # Using the residual of the model and original X_train for the test
    names = ['Lagrange multiplier statistic', 'p-value', 'f-value', 'f p-value']
    bp_test = het_breuschpagan(model_results.resid, model_results.model.exog)
    print(pd.Series(bp_test, index=names))
    results['breusch_pagan_test'] = pd.Series(bp_test, index=names).to_dict()

    # Variance Inflation Factor (VIF)
    print("\n--- Variance Inflation Factor (VIF) ---")
    X_train_no_const = X_train.drop(columns=['const'], errors='ignore') # Drop 'const' safely
    if not X_train_no_const.empty:
        vif_data = pd.DataFrame({
            'feature': X_train_no_const.columns,
            'VIF': [variance_inflation_factor(X_train_no_const.values, i) for i in range(X_train_no_const.shape[1])]
        })
        # Add 'const' back to VIF data for completeness, with NaN for VIF
        vif_data = pd.concat([pd.DataFrame([{'feature': 'const', 'VIF': np.nan}]), vif_data]).reset_index(drop=True)
    else:
        vif_data = pd.DataFrame([{'feature': 'const', 'VIF': np.nan}]) # Only const remains

    print(vif_data.sort_values(by="VIF", ascending=False))
    results['vif'] = vif_data.sort_values(by="VIF", ascending=False).to_dict('records')

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title('Test Set: Actual vs. Predicted Prices')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title('Test Set: Residuals vs. Predicted Prices')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']), reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')

            plt.tight_layout()
            plt.show()

    return results
````

#### **Evaluating the best OLS model (`df_transformed_with_new_price_extended_with_pca`) using the corrected function with robust standard errors to obtain its performance metrics and diagnostics.**



````python
print("\n--- Evaluating Model: df_transformed_with_new_price_extended_with_pca (Robust Standard Errors) ---")
ols_results_robust_best_model = evaluate_ols_model(df_transformed_with_new_price_extended_with_pca.copy(), 'log_Price', cov_type='HC3')
````

### **Implementing Ridge regression, including data scaling, hyperparameter tuning, and evaluation**



#### **Defining a function for Ridge regression:**

````python
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_regularized_model(
    df_input,
    target_col,
    model_type='ridge',
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95,
    param_grid=None
):
    """
    Performs Ridge or Lasso regression modeling, evaluates performance metrics,
    conducts hyperparameter tuning with K-fold Cross-Validation, and bootstrapping.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        model_type (str): Type of regularization model ('ridge' or 'lasso').
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results.
        param_grid (dict): Dictionary with parameters names (str) as keys and lists of
                           parameter settings to try as values for GridSearchCV.

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              and best model details.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for sklearn models
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Scale the features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Convert scaled arrays back to DataFrame for consistent column names (important for VIF or later interpretation)
    X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
    X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred):
        r2 = r2_score(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        mape_values = np.abs((y_true - y_pred) / y_true)
        mape_values = np.where(y_true == 0, np.nan, mape_values)
        mape = np.nanmean(mape_values) * 100

        return r2, mse, rmse, mae, mape

    # 4. Hyperparameter Tuning with GridSearchCV
    print(f"\n--- Hyperparameter Tuning for {model_type.capitalize()} Regression ---")
    if model_type == 'ridge':
        model_base = Ridge(random_state=random_state)
        if param_grid is None: # Default param_grid for Ridge if not provided
            param_grid = {'alpha': np.logspace(-4, 2, 50)}
    elif model_type == 'lasso':
        model_base = Lasso(random_state=random_state, max_iter=2000) # Increased max_iter for convergence
        if param_grid is None: # Default param_grid for Lasso if not provided
            param_grid = {'alpha': np.logspace(-4, 2, 50)}
    else:
        raise ValueError("model_type must be 'ridge' or 'lasso'")

    grid_search = GridSearchCV(
        model_base,
        param_grid,
        cv=n_splits_cv,
        scoring='neg_mean_squared_error',
        n_jobs=-1
    )
    grid_search.fit(X_train_scaled, y_train)

    best_model = grid_search.best_estimator_
    results['best_params'] = grid_search.best_params_
    print(f"Best parameters found: {grid_search.best_params_}")

    # 5. Evaluate Best Model Performance
    y_train_pred = best_model.predict(X_train_scaled)
    y_test_pred = best_model.predict(X_test_scaled)

    r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred)
    r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred)

    print("\n--- Best Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 6. K-fold Cross-Validation with the best model
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X_train_scaled_df)):
        X_train_fold, X_val_fold = X_train_scaled_df.iloc[train_index], X_train_scaled_df.iloc[val_index]
        y_train_fold, y_val_fold = y_train.iloc[train_index], y_train.iloc[val_index]

        model_fold = best_model # Use the best model found by GridSearchCV
        model_fold.fit(X_train_fold, y_train_fold) # Refit on current fold's training data
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold)
        cv_r2.append(r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))

    # 7. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae = [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train_scaled_df), len(X_train_scaled_df), replace=True)
        X_boot, y_boot = X_train_scaled_df.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit model on bootstrap sample
        model_boot = best_model # Start with the best model found by GridSearchCV
        model_boot.fit(X_boot, y_boot)

        # Predict on original test set (scaled)
        y_test_pred_boot = model_boot.predict(X_test_scaled)

        # Calculate and store metrics
        r2_boot, mse_boot, rmse_boot, mae_boot, mape_boot = calculate_metrics(y_test, y_test_pred_boot)
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)

        # Store coefficients (handling potential difference in number of features if some are dropped by Lasso)
        for idx, col in enumerate(X_train.columns):
            if model_type == 'lasso' and best_model.coef_[idx] == 0: # If Lasso shrinks coefficient to zero, store 0
                boot_coefficients[col].append(0)
            else:
                boot_coefficients[col].append(model_boot.coef_[idx])

    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE', 'MAPE'], [boot_r2, boot_mse, boot_rmse, boot_mae, boot_mape]):
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")
    if 'MAPE_mean' in metrics_ci and not np.isnan(metrics_ci['MAPE_mean']):
        print(f"Mean MAPE (Test): {metrics_ci['MAPE_mean']:.2f}% (95% CI: [{metrics_ci['MAPE_ci_lower']:.2f}%, {metrics_ci['MAPE_ci_upper']:.2f}%])")
    else:
        print("Mean MAPE (Test): Not available or could not be calculated due to NaN values.")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    # Display intercept separately
    print(f" {'(Intercept)':<25} Mean: {best_model.intercept_:.4f} (95% CI: [{best_model.intercept_:.4f}, {best_model.intercept_:.4f}])")
    for col, ci_vals in coefficients_ci.items():
        if col != 'const': # 'const' is not present in sklearn models
            print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title(f'Test Set: Actual vs. Predicted Prices ({model_type.capitalize()} Regression)')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title(f'Test Set: Residuals vs. Predicted Prices ({model_type.capitalize()} Regression)')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae, 'MAPE': boot_mape}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution', 'MAPE': 'Bootstrapped MAPE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1
        else:
            axes[plot_idx].set_title(f'No valid {metric_name} data')
            axes[plot_idx].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[plot_idx].transAxes)
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']) if not np.isnan(item[1]['mean']) else -1, reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')
                else:
                    axes[i].set_title(f'No valid {col_name} coeff data')
                    axes[i].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[i].transAxes)

            plt.tight_layout()
            plt.show()

    return results
````

#### **Updating the `evaluate_regularized_model` function to resolve the `boot_mape` initialization error.**



````python
from sklearn.linear_model import Ridge, Lasso
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV, KFold
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_ols_model(
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95,
    cov_type='nonrobust' # Added cov_type parameter with default
):
    """
    Performs OLS regression modeling, evaluates performance metrics, conducts OLS diagnostics,
    and visualizes results with K-fold Cross-Validation and Bootstrapping for confidence intervals.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results (e.g., 0.95 for 95% CI).
        cov_type (str): Type of covariance matrix estimator to use (e.g., 'nonrobust', 'HC0', 'HC1', 'HC3').

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              diagnostic results, and generated plots.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for statsmodels
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Add a constant to the features for OLS model
    X = sm.add_constant(X)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Fit an OLS model using statsmodels.api.OLS on the training data
    # Pass cov_type to the fit method
    model = sm.OLS(y_train, X_train)
    model_results = model.fit(cov_type=cov_type)

    # Obtain model predictions
    y_train_pred = model_results.predict(X_train)
    y_test_pred = model_results.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred, n_features):
        r2 = r2_score(y_true, y_pred)
        n = len(y_true)
        p = n_features - 1  # -1 because of the added constant
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        # MAPE calculation (handle division by zero if actual values are 0)
        # Ensure y_true does not contain zeros for MAPE calculation, or handle them gracefully.
        # For log-transformed prices, y_true should not be zero, but we add robustness.
        mape_values = np.abs((y_true - y_pred) / y_true)
        # Replace inf with nan if y_true contains zeros that lead to inf
        mape_values = np.where(y_true == 0, np.nan, mape_values)
        mape = np.nanmean(mape_values) * 100 # Use nanmean to ignore NaNs resulting from y_true=0

        return r2, adj_r2, mse, rmse, mae, mape

    # 4. Calculate and print metrics for training and testing sets
    n_features_train = X_train.shape[1]
    r2_train, adj_r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred, n_features_train)
    r2_test, adj_r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred, X_test.shape[1])

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train Adjusted R-squared: {adj_r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test Adjusted R-squared: {adj_r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'Adj_R2': adj_r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'Adj_R2': adj_r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 5. K-fold Cross-Validation
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_adj_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X)):
        X_train_fold, X_val_fold = X.iloc[train_index], X.iloc[val_index]
        y_train_fold, y_val_fold = y.iloc[train_index], y.iloc[val_index]

        # Pass cov_type to the fit method within CV loop
        model_fold = sm.OLS(y_train_fold, X_train_fold).fit(cov_type=cov_type)
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, adj_r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold, X_val_fold.shape[1])
        cv_r2.append(r2_fold)
        cv_adj_r2.append(adj_r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'Adj_R2_mean': np.mean(cv_adj_r2), 'Adj_R2_std': np.std(cv_adj_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))


    # 6. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae, boot_mape = [], [], [], [], [] # Initialize boot_mape
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train), len(X_train), replace=True)
        X_boot, y_boot = X_train.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit OLS model on bootstrap sample
        try:
            # Pass cov_type to the fit method within bootstrap loop
            model_boot = sm.OLS(y_boot, X_boot).fit(cov_type=cov_type)
        except Exception as e:
            # Handle cases where bootstrap sample might be degenerate
            # print(f"Warning: OLS failed for bootstrap sample {i}. Skipping. Error: {e}")
            continue

        # Predict on original test set
        y_test_pred_boot = model_boot.predict(X_test)

        # Calculate and store metrics
        r2_boot, _, mse_boot, rmse_boot, mae_boot, mape_boot = calculate_metrics(y_test, y_test_pred_boot, X_test.shape[1])
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)
        boot_mape.append(mape_boot) # Append mape_boot

        # Store coefficients
        for col in X_train.columns:
            if col in model_boot.params:
                boot_coefficients[col].append(model_boot.params[col])
            else:
                boot_coefficients[col].append(np.nan) # Append nan if coefficient not found (e.g., due to perfect multicollinearity in bootstrap sample)


    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE', 'MAPE'], [boot_r2, boot_mse, boot_rmse, boot_mae, boot_mape]): # Added MAPE
        # Filter out NaN values before calculating percentiles
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")
    if 'MAPE_mean' in metrics_ci and not np.isnan(metrics_ci['MAPE_mean']):
        print(f"Mean MAPE (Test): {metrics_ci['MAPE_mean']:.2f}% (95% CI: [{metrics_ci['MAPE_ci_lower']:.2f}%, {metrics_ci['MAPE_ci_upper']:.2f}%])")
    else:
        print("Mean MAPE (Test): Not available or could not be calculated due to NaN values.")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    for col, ci_vals in coefficients_ci.items():
        print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")


    # 7. OLS Diagnostics
    print("\n--- OLS Model Summary (Full Training Data) ---")
    print(model_results.summary()) # This summary will now use the specified cov_type
    results['ols_summary'] = model_results.summary().as_html()

    # Breusch-Pagan test for heteroskedasticity
    print("\n--- Breusch-Pagan Test for Heteroskedasticity ---")
    # The Breusch-Pagan test requires model residuals and the exogenous variables (X)
    # The test needs the original X_train, not X_train with constant for formula, but sm.add_constant() is standard.
    # Using the residual of the model and original X_train for the test
    names = ['Lagrange multiplier statistic', 'p-value', 'f-value', 'f p-value']
    bp_test = het_breuschpagan(model_results.resid, model_results.model.exog)
    print(pd.Series(bp_test, index=names))
    results['breusch_pagan_test'] = pd.Series(bp_test, index=names).to_dict()

    # Variance Inflation Factor (VIF)
    print("\n--- Variance Inflation Factor (VIF) ---")
    X_train_no_const = X_train.drop(columns=['const'], errors='ignore') # Drop 'const' safely
    if not X_train_no_const.empty:
        vif_data = pd.DataFrame({
            'feature': X_train_no_const.columns,
            'VIF': [variance_inflation_factor(X_train_no_const.values, i) for i in range(X_train_no_const.shape[1])]
        })
        # Add 'const' back to VIF data for completeness, with NaN for VIF
        vif_data = pd.concat([pd.DataFrame([{'feature': 'const', 'VIF': np.nan}]), vif_data]).reset_index(drop=True)
    else:
        vif_data = pd.DataFrame([{'feature': 'const', 'VIF': np.nan}]) # Only const remains

    print(vif_data.sort_values(by="VIF", ascending=False))
    results['vif'] = vif_data.sort_values(by="VIF", ascending=False).to_dict('records')

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title('Test Set: Actual vs. Predicted Prices')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title('Test Set: Residuals vs. Predicted Prices')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae, 'MAPE': boot_mape}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution', 'MAPE': 'Bootstrapped MAPE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1
        else:
            axes[plot_idx].set_title(f'No valid {metric_name} data')
            axes[plot_idx].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[plot_idx].transAxes)
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']) if not np.isnan(item[1]['mean']) else -1, reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')
                else:
                    axes[i].set_title(f'No valid {col_name} coeff data')
                    axes[i].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[i].transAxes)

            plt.tight_layout()
            plt.show()

    return results

def evaluate_regularized_model(
    df_input,
    target_col,
    model_type='ridge',
    test_size=0.2,
    random_state=42,
    n_splits_cv=5,
    n_bootstrap_samples=100,
    confidence_level=0.95,
    param_grid=None
):
    """
    Performs Ridge or Lasso regression modeling, evaluates performance metrics,
    conducts hyperparameter tuning with K-fold Cross-Validation, and bootstrapping.

    Args:
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        model_type (str): Type of regularization model ('ridge' or 'lasso').
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        n_splits_cv (int): Number of folds for K-fold Cross-Validation.
        n_bootstrap_samples (int): Number of bootstrap samples to generate.
        confidence_level (float): Confidence level for bootstrapping results.
        param_grid (dict): Dictionary with parameters names (str) as keys and lists of
                           parameter settings to try as values for GridSearchCV.

    Returns:
        dict: A dictionary containing all calculated metrics, confidence intervals,
              and best model details.
    """

    results = {}

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # 1. Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for sklearn models
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # 2. Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # 3. Scale the features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Convert scaled arrays back to DataFrame for consistent column names (important for VIF or later interpretation)
    X_train_scaled_df = pd.DataFrame(X_train_scaled, columns=X_train.columns, index=X_train.index)
    X_test_scaled_df = pd.DataFrame(X_test_scaled, columns=X_test.columns, index=X_test.index)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred):
        r2 = r2_score(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        mape_values = np.abs((y_true - y_pred) / y_true)
        mape_values = np.where(y_true == 0, np.nan, mape_values) # Handle division by zero
        mape = np.nanmean(mape_values) * 100

        return r2, mse, rmse, mae, mape

    # 4. Hyperparameter Tuning with GridSearchCV
    print(f"\n--- Hyperparameter Tuning for {model_type.capitalize()} Regression ---")
    if model_type == 'ridge':
        model_base = Ridge(random_state=random_state)
        if param_grid is None: # Default param_grid for Ridge if not provided
            param_grid = {'alpha': np.logspace(-4, 2, 50)}
    elif model_type == 'lasso':
        model_base = Lasso(random_state=random_state, max_iter=2000) # Increased max_iter for convergence
        if param_grid is None: # Default param_grid for Lasso if not provided
            param_grid = {'alpha': np.logspace(-4, 2, 50)}
    else:
        raise ValueError("model_type must be 'ridge' or 'lasso'")

    grid_search = GridSearchCV(
        model_base,
        param_grid,
        cv=n_splits_cv,
        scoring='neg_mean_squared_error',
        n_jobs=-1
    )
    grid_search.fit(X_train_scaled, y_train)

    best_model = grid_search.best_estimator_
    results['best_params'] = grid_search.best_params_
    print(f"Best parameters found: {grid_search.best_params_}")

    # 5. Evaluate Best Model Performance
    y_train_pred = best_model.predict(X_train_scaled)
    y_test_pred = best_model.predict(X_test_scaled)

    r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred)
    r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred)

    print("\n--- Best Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results['train_metrics'] = {'R2': r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train}
    results['test_metrics'] = {'R2': r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}

    # 6. K-fold Cross-Validation with the best model
    kf = KFold(n_splits=n_splits_cv, shuffle=True, random_state=random_state)
    cv_r2, cv_mse, cv_rmse, cv_mae, cv_mape = [], [], [], [], []

    print(f"\n--- K-fold Cross-Validation ({n_splits_cv} folds) ---")
    for fold, (train_index, val_index) in enumerate(kf.split(X_train_scaled_df)):
        X_train_fold, X_val_fold = X_train_scaled_df.iloc[train_index], X_train_scaled_df.iloc[val_index]
        y_train_fold, y_val_fold = y_train.iloc[train_index], y_train.iloc[val_index]

        model_fold = best_model # Use the best model found by GridSearchCV
        model_fold.fit(X_train_fold, y_train_fold) # Refit on current fold's training data
        y_val_pred_fold = model_fold.predict(X_val_fold)

        r2_fold, mse_fold, rmse_fold, mae_fold, mape_fold = calculate_metrics(y_val_fold, y_val_pred_fold)
        cv_r2.append(r2_fold)
        cv_mse.append(mse_fold)
        cv_rmse.append(rmse_fold)
        cv_mae.append(mae_fold)
        cv_mape.append(mape_fold)
        print(f"Fold {fold+1} - R2: {r2_fold:.4f}, RMSE: {rmse_fold:.4f}, MAE: {mae_fold:.4f}")

    results['cv_metrics'] = {
        'R2_mean': np.mean(cv_r2), 'R2_std': np.std(cv_r2),
        'MSE_mean': np.mean(cv_mse), 'MSE_std': np.std(cv_mse),
        'RMSE_mean': np.mean(cv_rmse), 'RMSE_std': np.std(cv_rmse),
        'MAE_mean': np.mean(cv_mae), 'MAE_std': np.std(cv_mae),
        'MAPE_mean': np.mean(cv_mape), 'MAPE_std': np.std(cv_mape),
    }
    print("\nCV Mean R-squared: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['R2_mean'], results['cv_metrics']['R2_std']))
    print("CV Mean RMSE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['RMSE_mean'], results['cv_metrics']['RMSE_std']))
    print("CV Mean MAE: {:.4f} (Std: {:.4f})".format(results['cv_metrics']['MAE_mean'], results['cv_metrics']['MAE_std']))
    print("CV Mean MAPE: {:.2f}% (Std: {:.2f}%)".format(results['cv_metrics']['MAPE_mean'], results['cv_metrics']['MAPE_std']))

    # 7. Bootstrapping
    boot_r2, boot_mse, boot_rmse, boot_mae, boot_mape = [], [], [], [], []
    boot_coefficients = {col: [] for col in X_train.columns}

    print(f"\n--- Bootstrapping ({n_bootstrap_samples} samples) ---")
    for i in range(n_bootstrap_samples):
        # Sample (with replacement) from the training data
        bootstrap_sample_indices = np.random.choice(len(X_train_scaled_df), len(X_train_scaled_df), replace=True)
        X_boot, y_boot = X_train_scaled_df.iloc[bootstrap_sample_indices], y_train.iloc[bootstrap_sample_indices]

        # Fit model on bootstrap sample
        model_boot = best_model # Start with the best model found by GridSearchCV
        model_boot.fit(X_boot, y_boot)

        # Predict on original test set (scaled)
        y_test_pred_boot = model_boot.predict(X_test_scaled)

        # Calculate and store metrics
        r2_boot, mse_boot, rmse_boot, mae_boot, mape_boot = calculate_metrics(y_test, y_test_pred_boot)
        boot_r2.append(r2_boot)
        boot_mse.append(mse_boot)
        boot_rmse.append(rmse_boot)
        boot_mae.append(mae_boot)
        boot_mape.append(mape_boot)

        # Store coefficients (handling potential difference in number of features if some are dropped by Lasso)
        for idx, col in enumerate(X_train.columns):
            if model_type == 'lasso' and best_model.coef_[idx] == 0: # If Lasso shrinks coefficient to zero, store 0
                boot_coefficients[col].append(0)
            else:
                boot_coefficients[col].append(model_boot.coef_[idx])

    # Calculate confidence intervals for metrics and coefficients
    alpha = (1 - confidence_level) / 2
    lower_percentile = alpha * 100
    upper_percentile = (1 - alpha) * 100

    metrics_ci = {}
    for metric_name, values in zip(['R2', 'MSE', 'RMSE', 'MAE', 'MAPE'], [boot_r2, boot_mse, boot_rmse, boot_mae, boot_mape]):
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            metrics_ci[metric_name + '_mean'] = np.mean(valid_values)
            metrics_ci[metric_name + '_ci_lower'] = np.percentile(valid_values, lower_percentile)
            metrics_ci[metric_name + '_ci_upper'] = np.percentile(valid_values, upper_percentile)
        else:
            metrics_ci[metric_name + '_mean'] = np.nan
            metrics_ci[metric_name + '_ci_lower'] = np.nan
            metrics_ci[metric_name + '_ci_upper'] = np.nan

    results['bootstrap_metrics_ci'] = metrics_ci

    print("\n--- Bootstrapping Confidence Intervals ---")
    print(f"Mean R-squared (Test): {metrics_ci['R2_mean']:.4f} (95% CI: [{metrics_ci['R2_ci_lower']:.4f}, {metrics_ci['R2_ci_upper']:.4f}])")
    print(f"Mean MSE (Test): {metrics_ci['MSE_mean']:.4f} (95% CI: [{metrics_ci['MSE_ci_lower']:.4f}, {metrics_ci['MSE_ci_upper']:.4f}])")
    print(f"Mean RMSE (Test): {metrics_ci['RMSE_mean']:.4f} (95% CI: [{metrics_ci['RMSE_ci_lower']:.4f}, {metrics_ci['RMSE_ci_upper']:.4f}])")
    print(f"Mean MAE (Test): {metrics_ci['MAE_mean']:.4f} (95% CI: [{metrics_ci['MAE_ci_lower']:.4f}, {metrics_ci['MAE_ci_upper']:.4f}])")
    if 'MAPE_mean' in metrics_ci and not np.isnan(metrics_ci['MAPE_mean']):
        print(f"Mean MAPE (Test): {metrics_ci['MAPE_mean']:.2f}% (95% CI: [{metrics_ci['MAPE_ci_lower']:.2f}%, {metrics_ci['MAPE_ci_upper']:.2f}%])")
    else:
        print("Mean MAPE (Test): Not available or could not be calculated due to NaN values.")

    coefficients_ci = {}
    for col, values in boot_coefficients.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            coefficients_ci[col] = {
                'mean': np.mean(valid_values),
                'ci_lower': np.percentile(valid_values, lower_percentile),
                'ci_upper': np.percentile(valid_values, upper_percentile)
            }
        else:
            coefficients_ci[col] = {'mean': np.nan, 'ci_lower': np.nan, 'ci_upper': np.nan}

    results['bootstrap_coefficients_ci'] = coefficients_ci

    print("\n--- Bootstrapping Coefficient Confidence Intervals ---")
    # Display intercept separately
    print(f" {(len('(Intercept)') + 1) * ' ':<25} Mean: {best_model.intercept_:.4f} (95% CI: [{best_model.intercept_:.4f}, {best_model.intercept_:.4f}])")
    for col, ci_vals in coefficients_ci.items():
        if col != 'const': # 'const' is not present in sklearn models
            print(f" {col:<25} Mean: {ci_vals['mean']:.4f} (95% CI: [{ci_vals['ci_lower']:.4f}, {ci_vals['ci_upper']:.4f}])")

    # 8. Visualizations
    results['plots'] = {}

    # Actual vs. Predicted values for the test set
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test, y=y_test_pred, alpha=0.6)
    plt.plot([min(y_test), max(y_test)], [min(y_test), max(y_test)], color='red', linestyle='--') # 45 degree line
    plt.title(f'Test Set: Actual vs. Predicted Prices ({model_type.capitalize()} Regression)')
    plt.xlabel('Actual Price (log)')
    plt.ylabel('Predicted Price (log)')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Residuals vs. Predicted values for the test set
    residuals = y_test - y_test_pred
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.6)
    plt.axhline(y=0, color='red', linestyle='--')
    plt.title(f'Test Set: Residuals vs. Predicted Prices ({model_type.capitalize()} Regression)')
    plt.xlabel('Predicted Price (log)')
    plt.ylabel('Residuals')
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    # Plot histograms or kernel density estimates of the bootstrapped metrics distributions
    fig, axes = plt.subplots(1, 4, figsize=(20, 5))
    metrics_to_plot = {'R2': boot_r2, 'RMSE': boot_rmse, 'MAE': boot_mae, 'MAPE': boot_mape}
    metric_titles = {'R2': 'Bootstrapped R-squared Distribution', 'RMSE': 'Bootstrapped RMSE Distribution', 'MAE': 'Bootstrapped MAE Distribution', 'MAPE': 'Bootstrapped MAPE Distribution'}

    plot_idx = 0
    for metric_name, values in metrics_to_plot.items():
        valid_values = [v for v in values if not np.isnan(v)]
        if valid_values:
            sns.histplot(valid_values, kde=True, ax=axes[plot_idx])
            axes[plot_idx].set_title(metric_titles[metric_name])
            axes[plot_idx].set_xlabel(metric_name)
            axes[plot_idx].set_ylabel('Density')
            plot_idx += 1
        else:
            axes[plot_idx].set_title(f'No valid {metric_name} data')
            axes[plot_idx].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[plot_idx].transAxes)
            plot_idx += 1

    plt.tight_layout()
    plt.show()

    # For a few key coefficients, plot histograms or kernel density estimates of their bootstrapped distributions.
    # Select top 5 coefficients by absolute mean value or just a few interesting ones
    if boot_coefficients:
        sorted_coeffs = sorted(coefficients_ci.items(), key=lambda item: abs(item[1]['mean']) if not np.isnan(item[1]['mean']) else -1, reverse=True)
        top_n_coeffs_names = [item[0] for item in sorted_coeffs if item[0] != 'const'][:5] # Exclude 'const' and take top 5

        if top_n_coeffs_names:
            fig, axes = plt.subplots(1, min(len(top_n_coeffs_names), 5), figsize=(5 * min(len(top_n_coeffs_names), 5), 5))
            if not isinstance(axes, np.ndarray): # Handle case where only one subplot is created
                axes = [axes]

            for i, col_name in enumerate(top_n_coeffs_names):
                values = boot_coefficients[col_name]
                valid_values = [v for v in values if not np.isnan(v)]
                if valid_values:
                    sns.histplot(valid_values, kde=True, ax=axes[i])
                    axes[i].set_title(f'Bootstrapped {col_name} Coeff')
                    axes[i].set_xlabel(f'{col_name} Coefficient')
                    axes[i].set_ylabel('Density')
                else:
                    axes[i].set_title(f'No valid {col_name} coeff data')
                    axes[i].text(0.5, 0.5, 'N/A', horizontalalignment='center', verticalalignment='center', transform=axes[i].transAxes)

            plt.tight_layout()
            plt.show()

    return results
````

## **Comparing Regularized Models with Best OLS Models:**

 * Comparing the performance of the Ridge and Lasso models against the best OLS model, analyzing the impact of regularization on predictive accuracy, coefficient stability, and interpretability, especially concerning multicollinearity.


#### **Collecting the key performance metrics from all the regularized models to create a comparative summary table:**







````python
metrics_summary = pd.DataFrame(columns=[
    'Model',
    'Test_R2', 'Test_RMSE', 'Test_MAE',
    'CV_Mean_R2', 'CV_R2_Std',
    'CV_Mean_RMSE', 'CV_RMSE_Std',
    'CV_Mean_MAE', 'CV_MAE_Std',
    'Boot_Mean_R2', 'Boot_R2_CI_Lower', 'Boot_R2_CI_Upper',
    'Boot_Mean_RMSE', 'Boot_RMSE_CI_Lower', 'Boot_RMSE_CI_Upper',
    'Boot_Mean_MAE', 'Boot_MAE_CI_Lower', 'Boot_MAE_CI_Upper'

])

def add_model_metrics(model_name, results_dict):
    global metrics_summary
    new_row = {
        'Model': model_name,
        'Test_R2': results_dict['test_metrics']['R2'],
        'Test_RMSE': results_dict['test_metrics']['RMSE'],
        'Test_MAE': results_dict['test_metrics']['MAE'],
        'CV_Mean_R2': results_dict['cv_metrics']['R2_mean'],
        'CV_R2_Std': results_dict['cv_metrics']['R2_std'],
        'CV_Mean_RMSE': results_dict['cv_metrics']['RMSE_mean'],
        'CV_RMSE_Std': results_dict['cv_metrics']['RMSE_std'],
        'CV_Mean_MAE': results_dict['cv_metrics']['MAE_mean'],
        'CV_MAE_Std': results_dict['cv_metrics']['MAE_std'],
        'Boot_Mean_R2': results_dict['bootstrap_metrics_ci']['R2_mean'],
        'Boot_R2_CI_Lower': results_dict['bootstrap_metrics_ci']['R2_ci_lower'],
        'Boot_R2_CI_Upper': results_dict['bootstrap_metrics_ci']['R2_ci_upper'],
        'Boot_Mean_RMSE': results_dict['bootstrap_metrics_ci']['RMSE_mean'],
        'Boot_RMSE_CI_Lower': results_dict['bootstrap_metrics_ci']['RMSE_ci_lower'],
        'Boot_RMSE_CI_Upper': results_dict['bootstrap_metrics_ci']['RMSE_ci_upper'],
        'Boot_Mean_MAE': results_dict['bootstrap_metrics_ci']['MAE_mean'],
        'Boot_MAE_CI_Lower': results_dict['bootstrap_metrics_ci']['MAE_ci_lower'],
        'Boot_MAE_CI_Upper': results_dict['bootstrap_metrics_ci']['MAE_ci_upper']

    }
    metrics_summary = pd.concat([metrics_summary, pd.DataFrame([new_row])], ignore_index=True)

# Add metrics for each model
add_model_metrics('OLS_Robust_Best', ols_results_robust_best_model)
add_model_metrics('Ridge_Best_Model', ridge_results_best_model)
add_model_metrics('Lasso_Best_Model', lasso_results_best_model)
add_model_metrics('Ridge_No_NewPrice', ridge_results_no_new_price_model)
add_model_metrics('Lasso_No_NewPrice', lasso_results_no_new_price_model)

# Sort the DataFrame by Test_R2 in descending order
metrics_summary_sorted = metrics_summary.sort_values(by='Test_R2', ascending=False).reset_index(drop=True)

# Function to bold the best value for each metric
def highlight_best_linear(s):
    if s.name in ['Test_R2', 'Boot_Mean_R2', 'CV_Mean_R2']:
        is_best = s == s.max()
    elif s.name in ['Test_RMSE', 'Test_MAE', 'CV_Mean_RMSE', 'CV_Mean_MAE', 'Boot_Mean_RMSE', 'Boot_Mean_MAE']:
        is_best = s == s.min()
    else:
        return [''] * len(s)
    return ['font-weight: bold' if v else '' for v in is_best]

# Display the summary table with highlighting
print("\n--- Model Performance Summary (Sorted by Test R2) ---")
display(metrics_summary_sorted.style.apply(highlight_best_linear, subset=pd.Index(metrics_summary_sorted.columns).drop('Model')))
````

* **Cross-Validation Stability:** The OLS_Robust_Best model shows slightly better stability in cross-validation metrics with a higher CV_Mean_R2 (0.9479) and lower CV_R2_Std (0.0055) compared to Ridge_Best_Model and Lasso_Best_Model (both with CV_Mean_R2 around 0.9458 and CV_R2_Std around 0.008).

*   **Bootstrap Stability**: The bootstrap confidence intervals for all models are relatively tight, suggesting good stability in their performance estimates across different resamples of the data.

```
 Location_Bangalore        Mean: 0.0241 (95% CI: [0.0154, 0.0318])
 Location_Chennai          Mean: -0.0020 (95% CI: [-0.0114, 0.0039])
 ... (many more categorical features, some coefficients driven to zero)
```

**Observations:**

*   **Coefficient Shrinkage (Ridge)**: Ridge regression effectively shrinks coefficients towards zero, as expected. Many coefficients for individual categorical features (e.g., specific brands or locations) are smaller in magnitude compared to OLS, and their confidence intervals are tighter, indicating more stable estimates. However, Ridge does not set coefficients exactly to zero.
*   **Feature Selection (Lasso)**: Lasso regression's primary advantage is feature selection. Many coefficients for various brand and seat categories (e.g., `Brand_Bentley`, `Brand_Chevrolet`, `Brand_Force`, `Brand_ISUZU`, `Brand_Lamborghini`, `Brand_Porsche`, `Seats_9.0`, `Seats_10.0`) have been driven exactly to zero. This is a powerful outcome, as it indicates that these features were deemed less important or redundant for predictive power by the Lasso model, leading to a sparser and potentially more interpretable model. For other features, the coefficients are generally smaller than OLS, similar to Ridge.
*   **Stability of Important Predictors**: Across all three models with `log_New_price`, coefficients for highly influential features like `log_New_price`, `Power_Engine_PCA`, `log_New_price_x_Car_Age_Interaction`, `log_Kilometers_Driven_x_Car_Age_Interaction`, and `Kilometers_per_Year` remain significant and have relatively stable confidence intervals, even if their magnitudes are slightly adjusted by regularization. This confirms their consistent importance.

### **3. Multicollinearity (VIF) and Heteroskedasticity**

#### **VIF Values (OLS_Robust_Best model - similar trends for others):**
```
                                        feature          VIF
45                                 Brand_Maruti  1347.346802
37                                Brand_Hyundai   777.274203
36                                  Brand_Honda   527.835377
29                                    Brand_BMW   344.111923
6   log_Kilometers_Driven_x_Car_Age_Interaction   317.771184
... (many other high VIFs for Brand/Seats)
```

**Observations:**

*   **Persistent Multicollinearity for One-Hot Encoded Features**: Despite PCA for `log_Engine`/`log_Power` (where `Power_Engine_PCA` has a manageable VIF of around 5-6), the VIF values for many one-hot encoded `Brand` and `Seats` categories remain extremely high or infinite in the OLS model. This indicates severe multicollinearity within these categorical features, likely due to sparse categories and the nature of one-hot encoding without dropping a reference category or with very few observations in some categories that are not being dropped by the model. Ridge and Lasso, by their nature, are designed to handle multicollinearity by shrinking coefficients, which implicitly addresses this issue by reducing the impact of highly correlated predictors.
*   **Robust Standard Errors for OLS**: The use of `cov_type='HC3'` in the OLS model successfully generated robust standard errors. While the *coefficients* themselves remained unchanged, the *p-values* and *confidence intervals* for some coefficients shifted. This is evident from the OLS summary, where some previously significant features (based on non-robust errors) might become non-significant, and vice-versa, offering a more reliable assessment of statistical significance in the presence of heteroskedasticity.
*   **Heteroskedasticity Remains**: The Breusch-Pagan test still indicates significant heteroskedasticity in the robust OLS model (p-value = 8.44e-09). This means that while robust standard errors provide valid statistical inference, the underlying issue of non-constant error variance persists. Ridge and Lasso do not directly address heteroskedasticity; they primarily target multicollinearity and overfitting.

### **4. Impact of Regularization on Predictive Accuracy, Stability, and Interpretability**

**Predictive Accuracy:**
*   **Best Model Group**: OLS_Robust_Best, Ridge_Best_Model, and Lasso_Best_Model (all with `log_New_price` and extended features) achieve the highest predictive accuracy with very similar metrics. Regularization (Ridge/Lasso) does not offer a significant improvement in R-squared, RMSE, or MAE over OLS when the full, feature-engineered dataset (including `log_New_price`) is used. This suggests that the OLS model, despite its diagnostic issues, is already performing near the optimal predictive capacity for this particular feature set.
*   **No New Price Group**: Ridge_No_NewPrice and Lasso_No_NewPrice demonstrate that without `log_New_price`, accuracy drops considerably, and regularization does not recover this loss. Their performance is also very similar to each other.

**Stability:**
*   All models demonstrate good stability, as indicated by the tight bootstrap confidence intervals for their performance metrics. The standard deviations for CV metrics are also low. Regularization helps ensure this stability, especially in the presence of multicollinearity, by providing more reliable coefficient estimates.

**Interpretability:**
*   **OLS_Robust_Best**: Provides direct interpretation of coefficient magnitudes and signs. With robust standard errors, the p-values for coefficients become more trustworthy, allowing for more confident statements about which features significantly impact price. However, severe multicollinearity for categorical features (high VIFs) can still make the individual interpretation of highly correlated coefficients difficult.
*   **Ridge_Best_Model**: Improves coefficient stability and reduces their magnitudes but retains all features. Interpretation remains similar to OLS, but with a reduced risk of wildly fluctuating coefficients due to multicollinearity.
*   **Lasso_Best_Model**: Significantly enhances interpretability through feature selection, driving many less important coefficients to zero. This creates a sparser model, highlighting the most influential features explicitly. For example, if certain less popular brands or seat configurations have their coefficients set to zero, it simplifies understanding which brands/seats truly matter in the model. This makes Lasso appealing for models where parsimony and clear identification of key drivers are desired.

### **2. Statistically Significant Variables with Strong Predictive Power**

The variables listed below consistently show strong statistical significance (low p-values in OLS, or confidence intervals not crossing zero in regularized models) and substantial impact (higher absolute coefficient magnitudes) across the top 3 models. The inclusion of `log_New_price` and its interactions is paramount.

#### **Numerical and Engineered Features:**

*   **`log_New_price`**: This is by far the **most influential predictor**. Its coefficient (around 0.34-0.43) is consistently high and highly significant across all models, confirming that a car's new price is the strongest determinant of its used price.
*   **`log_New_price_x_Car_Age_Interaction`**: This interaction term is also highly significant and has a notable negative coefficient (around -0.017 to -0.12), indicating that the depreciation rate (how quickly a car's value drops) is influenced by its new price and age.
*   **`Power_Engine_PCA`**: The principal component combining `log_Power` and `log_Engine` is a very strong positive predictor (coefficient around 0.27-0.62). This implies that cars with higher engine power and displacement generally command higher used prices.
*   **`Kilometers_per_Year`**: This feature (coefficient around -0.000038 to -0.00005) is consistently significant and negatively impacts price. Cars driven more kilometers per year tend to depreciate faster.
*   **`Mileage`**: The fuel efficiency (Mileage) is a significant negative predictor (coefficient around -0.007 to -0.028). Lower mileage (better fuel efficiency) is generally associated with higher prices.
*   **`log_Kilometers_Driven_x_Car_Age_Interaction`**: This interaction term is significant and negatively impacts price (coefficient around -0.014 to -0.33), suggesting that the combined effect of high kilometers driven and older age further reduces the car's value.
*   **`Is_Premium_New_Price`**: This binary indicator (coefficient around -0.017 to -0.044) is significant, implying a distinct pricing dynamic for cars originally in the premium segment.
*   **`Car_Age`**: While `Car_Age` alone can be borderline significant, its impact is largely captured by interaction terms and `Kilometers_per_Year`.

#### **Categorical Features (One-Hot Encoded):**

*   **`Brand`**: Many brand-specific dummy variables show strong significance. Premium brands (e.g., `Brand_Audi`, `Brand_BMW`, `Brand_Jaguar`, `Brand_Land`, `Brand_Mercedes-Benz`, `Brand_Mini`, `Brand_Volvo`) consistently have positive coefficients, indicating higher used prices relative to the baseline brand. Conversely, some mass-market/economy brands (e.g., `Brand_Tata`, `Brand_Chevrolet`, `Brand_Hyundai`, `Brand_Maruti`, `Brand_Volkswagen`, `Brand_Nissan`, `Brand_Ford`) have negative coefficients. Lasso explicitly sets several less common brand coefficients to zero (e.g., `Brand_Bentley`, `Brand_Chevrolet`, `Brand_Force`, `Brand_ISUZU`, `Brand_Lamborghini`, `Brand_Porsche`), simplifying the model by indicating they have no significant unique effect after accounting for other features.
*   **`Location`**: Several location-specific dummy variables are significant (e.g., `Location_Bangalore`, `Location_Coimbatore`, `Location_Delhi`, `Location_Hyderabad`, `Location_Kolkata`, `Location_Mumbai`, `Location_Pune`), indicating regional price variations. For instance, `Location_Kolkata` consistently has a notable negative coefficient, suggesting lower prices there relative to the baseline location (Ahmedabad).
*   **`Transmission_Manual`**: This feature is consistently significant with a negative coefficient (around -0.013 to -0.03), implying that manual transmission cars are generally priced lower than automatic ones.
*   **`Fuel_Type_Grouped_Petrol`**: This is a significant negative predictor (coefficient around -0.05 to -0.11), indicating that petrol cars are generally priced lower than diesel cars (the baseline for fuel type).
*   **`Owner_Type_Revised_Second`**: This is a significant negative predictor (coefficient around -0.016 to -0.05), implying that second-owner cars are cheaper than first-owner cars.
*   **`Owner_Type_Revised_Multiple`**: This grouped category for multiple previous owners also shows a negative coefficient (around -0.011 to -0.16) and is significant, further emphasizing depreciation with more owners.
*   **`Seats`**: Most `Seats` categories (4.0 to 8.0) are not consistently significant across all models, and Lasso often drives their coefficients to zero, suggesting that the number of seats itself might not be a strong standalone predictor when other factors (like brand and new price) are accounted for, or its effect is absorbed by other highly correlated variables.


**Impact of `Electric` Fuel Type on Used Car Price**: In the OLS_Robust_Best model, the coefficient for `Fuel_Type_Grouped_Electric` was approximately 1.9091, but its p-value was 0.274, indicating no statistically significant evidence of a unique impact on log_Price compared to the baseline fuel type (Diesel), after accounting for other factors. The bootstrapped 95% confidence interval for this coefficient [-0.0000, 2.1678] also crossed zero, reinforcing this lack of significance. This lack of statistical significance is very likely due to the extremely limited number of `Electric` car observations (only 2 in the dataset). With such a small sample size, the model lacks sufficient data to confidently estimate a unique effect, leading to high uncertainty (wide confidence intervals) and thus, high p-values. Even if there's a real effect (which is theoretically expected), the model cannot statistically confirm it with so few examples.

## **Summary of OLS Regression Modeling:**

### **Analysis and Key Findings:**

* **Impact of log_New_price on Predictive Accuracy:** Models incorporating the `log_New_price` feature significantly outperform those that do not. For instance, models with `log_New_price` achieved a Test R-squared of approximately 0.95 and a Test MAPE of around 8.15%, whereas models without this feature saw their Test R-squared drop to about 0.94 and their Test MAPE increase substantially to approximately 15.31%. This highlights `log_New_price` as a crucial predictor for used car prices.
* **Heteroskedasticity Persistence:** The best OLS model, even after re-evaluation with robust standard errors, still exhibited significant heteroskedasticity (Breusch-Pagan test p-value: 8.44e-09). While robust standard errors corrected the statistical inference (p-values and confidence intervals), the underlying non-constant error variance in the residuals was not eliminated.
* **Multicollinearity in OLS Models:** High Variance Inflation Factor (VIF) values persisted for several one-hot encoded categorical features (e.g., Brand_Maruti with VIF 1347.35, Brand_Hyundai with VIF 777.27) in the OLS model, indicating severe multicollinearity.
* **Similar Predictive Performance for Top Models:** When the comprehensive feature set (including `log_New_price`) was used, the Robust OLS, Ridge, and Lasso models exhibited nearly identical predictive performance. For example, their Test R-squared values were all around 0.95, Test RMSE around 0.175, Test MAE around 0.134, and Test MAPE around 8.15%. This suggests that for this rich feature set, the OLS model's predictive capability was already near optimal, and regularization did not offer significant gains in accuracy.
* **Regularization Benefits in Interpretability and Stability:**
  * Ridge Regression effectively shrunk coefficient magnitudes, contributing to more stable estimates in the presence of multicollinearity, but it did not eliminate any features.
  * Lasso Regression demonstrated significant feature selection by driving many coefficients of less important or redundant features (e.g., specific brands like 'Bentley', 'Chevrolet', or 'Force', and certain seat configurations) exactly to zero. This created a sparser and more interpretable model without sacrificing predictive accuracy.
* **Model Stability:** Cross-validation analysis indicated that all evaluated models displayed good stability, with relatively tight standard deviations for their performance metrics across folds, implying consistent performance across different data samples.


### **Best OLS Model Identification and Justification:**

Both the `Ridge_Best_Model` and `OLS_Robust_Best` models demonstrate the highest R-squared (0.95), and the lowest RMSE (0.175), MAE (0.134), and MAPE (8.15%). This indicates they are the top performers in terms of predictive accuracy on unseen data. Comparing their cross-validation results, the `OLS_Robust_Best` model exhibits a slightly better CV_Mean_R2 (0.95 vs. 0.95) and CV_Mean_RMSE (0.172 vs. 0.172), suggesting marginally better and more stable performance across cross-validation folds in terms of overall variance explained and prediction error magnitude. While `Ridge_Best_Model` shows slightly lower CV_Mean_MAE (0.130 vs. 0.131) and CV_Mean_MAPE (7.50% vs. 7.63%), the robust OLS model's advantages in CV_Mean_R2 and CV_Mean_RMSE are considered more significant indicators of overall model fit and prediction accuracy.

**Based on the test and cross-validation performance metrics, the `OLS_Robust_Best` model stands out as the best performer among the linear and regularized models.** While Ridge and Lasso also achieve excellent predictive accuracy, the OLS_Robust_Best model exhibits a slightly better CV_Mean_R2 (0.95 vs. 0.95) and CV_Mean_RMSE (0.172 vs. 0.172). This indicates marginally superior and more consistent performance across cross-validation folds in terms of overall variance explained and prediction error magnitude.

Multicollinearity, as evidenced by high VIF values, was present in the dataset. While regularization techniques like Ridge and Lasso do not eliminate the underlying correlations between features, they **mitigate the negative effects of multicollinearity** on model stability by shrinking or zeroing out coefficients. However, the `OLS_Robust_Best` model, despite the inherent multicollinearity in the data, demonstrated comparable or slightly superior predictive accuracy and stability. **A key advantage of the OLS model, particularly when using robust standard errors, is the direct interpretability of its unbiased coefficient estimates.** Unlike regularized models, whose coefficients are biased due to the penalty term, OLS provides the best linear unbiased estimates. By employing robust standard errors, the statistical inferences (p-values and confidence intervals) regarding these unbiased coefficients remain valid, even in the presence of heteroskedasticity. **This combination of top-tier predictive accuracy, superior cross-validation stability, and reliable interpretability of unbiased coefficients makes the `OLS_Robust_Best model` the most suitable choice for this project among the linear regression approaches.**




## **Implementing Tree-based Models**

### **Tree-based Model Evaluation Strategy**

To further enhance the predictive modeling for used car prices, several tree-based regression models will be implemented and evaluated. This approach allows for the exploration of non-linear relationships and potentially capture more complex patterns in the data that linear models might miss. The strategy for evaluating these models is as follows:

#### **1. Chosen Model Types:**
A diverse set of tree-based models will be evaluated, which are known for their strong performance in regression tasks:
*   **Decision Tree Regressor**
*   **Random Forest Regressor**
*   **Gradient Boosting Regressor**
*   **XGBoost Regressor (Extreme Gradient Boosting)**
*   **LightGBM Regressor (Light Gradient Boosting Machine)**
*   **CatBoost Regressor (Categorical Boosting)**

### **2. Approach for Initial Screening with Selected DataFrames:**
Given the computational resources and time available, an initial screening approach will be used. Two key DataFrames will be selected that represent different aspects of the feature engineering efforts:

*   **`df_transformed_no_new_price_extended_with_pca`**: This DataFrame represents a comprehensive set of engineered features, including log-transformed numerical variables, interaction terms, squared terms, `Kilometers_per_Year`, and `Power_Engine_PCA`, but importantly, **excludes the `log_New_price` feature**. This wil allow for evaluation of model performance on the larger dataset (with 5897 observations).

*   **`df_transformed_with_new_price_extended_with_pca`**: This DataFrame includes all the features from the previous one, plus the highly impactful `log_New_price` feature and its related interaction terms. This model operates on a smaller, but potentially more predictive, dataset (with 2644 observations).

By comparing the performance of tree-based models on these two DataFrames, the trade-off between dataset size and the predictive power gained from including `log_New_price` in the context of non-linear models can be understood.

### **3. Evaluation Metrics and Constraints:**

Due to the extensive computational requirements observed during the evaluation of OLS models, and to manage overall execution time, **extensive K-fold Cross-Validation and Bootstrapping will be omitted from the evaluation of tree-based models** for initial screening. For the initial evaluation phase, the primary focus will be solely on **train/test set performance metrics**. This includes:

*   **R-squared**
*   **Mean Squared Error (MSE)**
*   **Root Mean Squared Error (RMSE)**
*   **Mean Absolute Error (MAE)**
*   **Mean Absolute Percentage Error (MAPE)**

This streamlined evaluation approach will allow for a quicker assessment of the relative performance of various tree-based models on the selected feature sets. The models demonstrating the most promising train/test performance will then be selected for more rigorous fine-tuning and validation, potentially incorporating more advanced cross-validation strategies if justified by their initial results and available resources.

#### **Importing all the necessary tree-based models along with other relevant libraries:**



````python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from catboost import CatBoostRegressor

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

print("Tree-based model libraries imported successfully.")
````

#### **Defining a function to evaluate the tree-based models:**



````python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

def evaluate_tree_model(
    model,
    model_name,
    df_input,
    target_col,
    test_size=0.2,
    random_state=42,
    cat_features_indices=None
):
    """
    Evaluates a given tree-based regression model on a specified DataFrame.

    Args:
        model: An unfitted sklearn or compatible regression model instance.
        model_name (str): Name of the model for printing results.
        df_input (pd.DataFrame): The input DataFrame containing features and target.
        target_col (str): The name of the target variable column.
        test_size (float): The proportion of the dataset to include in the test split.
        random_state (int): Seed for random number generation for reproducibility.
        cat_features_indices (list): List of indices for categorical features (for CatBoost).

    Returns:
        dict: A dictionary containing performance metrics for train and test sets.
    """

    print(f"\n--- Evaluating {model_name} ---")

    # Ensure no inf or nan values remain in the input DataFrame before any processing
    initial_rows_count = len(df_input)
    df_input.replace([np.inf, -np.inf], np.nan, inplace=True)
    df_input.dropna(inplace=True)
    if len(df_input) < initial_rows_count:
        print(f"Warning: Dropped {initial_rows_count - len(df_input)} rows from the input DataFrame due to inf/nan values.")

    # Separate features (X) from target (y)
    X = df_input.drop(columns=[target_col])
    y = df_input[target_col]

    # Convert boolean columns to int type for sklearn models
    for col in X.select_dtypes(include='bool').columns:
        X[col] = X[col].astype(int)

    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

    # Fit the model
    if model_name == 'CatBoost Regressor':
        model.fit(X_train, y_train, verbose=False, cat_features=cat_features_indices)
    else:
        model.fit(X_train, y_train)

    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    # Helper function to calculate metrics
    def calculate_metrics(y_true, y_pred):
        r2 = r2_score(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        mae = mean_absolute_error(y_true, y_pred)

        mape_values = np.abs((y_true - y_pred) / y_true)
        mape_values = np.where(y_true == 0, np.nan, mape_values) # Handle division by zero
        mape = np.nanmean(mape_values) * 100

        return r2, mse, rmse, mae, mape

    # Calculate metrics for training and testing sets
    r2_train, mse_train, rmse_train, mae_train, mape_train = calculate_metrics(y_train, y_train_pred)
    r2_test, mse_test, rmse_test, mae_test, mape_test = calculate_metrics(y_test, y_test_pred)

    print("\n--- Model Performance (Train/Test) ---")
    print(f"Train R-squared: {r2_train:.4f}")
    print(f"Train MSE: {mse_train:.4f}")
    print(f"Train RMSE: {rmse_train:.4f}")
    print(f"Train MAE: {mae_train:.4f}")
    print(f"Train MAPE: {mape_train:.2f}%")
    print("\n")
    print(f"Test R-squared: {r2_test:.4f}")
    print(f"Test MSE: {mse_test:.4f}")
    print(f"Test RMSE: {rmse_test:.4f}")
    print(f"Test MAE: {mae_test:.4f}")
    print(f"Test MAPE: {mape_test:.2f}%")

    results = {
        'model': model,
        'train_metrics': {'R2': r2_train, 'MSE': mse_train, 'RMSE': rmse_train, 'MAE': mae_train, 'MAPE': mape_train},
        'test_metrics': {'R2': r2_test, 'MSE': mse_test, 'RMSE': rmse_test, 'MAE': mae_test, 'MAPE': mape_test}
    }
    return results

print("Tree model evaluation function defined successfully.")
````

#### **Evaluating the Decision Tree Regressor on the `df_transformed_with_new_price_extended_with_pca` DataFrame**



````python
print("\n--- Evaluating DecisionTreeRegressor on df_transformed_with_new_price_extended_with_pca ---")
dtr = DecisionTreeRegressor(random_state=42)
dtr_results_with_new_price = evaluate_tree_model(
    model=dtr,
    model_name="DecisionTree Regressor",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price"
)
````

#### **Summary Table of Tree-based Models Performance Metrics:**



````python
tree_metrics_summary = pd.DataFrame(columns=[
    'Model', 'DataFrame',
    'Train_R2', 'Train_RMSE', 'Train_MAE', 'Train_MAPE',
    'Test_R2', 'Test_RMSE', 'Test_MAE', 'Test_MAPE'
])

def add_tree_model_metrics(model_name, dataframe_name, results_dict):
    global tree_metrics_summary
    new_row = {
        'Model': model_name,
        'DataFrame': dataframe_name,
        'Train_R2': results_dict['train_metrics']['R2'],
        'Train_RMSE': results_dict['train_metrics']['RMSE'],
        'Train_MAE': results_dict['train_metrics']['MAE'],
        'Train_MAPE': results_dict['train_metrics']['MAPE'],
        'Test_R2': results_dict['test_metrics']['R2'],
        'Test_RMSE': results_dict['test_metrics']['RMSE'],
        'Test_MAE': results_dict['test_metrics']['MAE'],
        'Test_MAPE': results_dict['test_metrics']['MAPE']
    }
    tree_metrics_summary = pd.concat([tree_metrics_summary, pd.DataFrame([new_row])], ignore_index=True)

# Add metrics for models evaluated on df_transformed_with_new_price_extended_with_pca
add_tree_model_metrics('DecisionTree Regressor', 'With New Price', dtr_results_with_new_price)
add_tree_model_metrics('RandomForest Regressor', 'With New Price', rfr_results_with_new_price)
add_tree_model_metrics('GradientBoosting Regressor', 'With New Price', gbr_results_with_new_price)
add_tree_model_metrics('XGBoost Regressor', 'With New Price', xgb_results_with_new_price)
add_tree_model_metrics('LightGBM Regressor', 'With New Price', lgbm_results_with_new_price)
add_tree_model_metrics('CatBoost Regressor', 'With New Price', cbr_results_with_new_price)

# Add metrics for models evaluated on df_transformed_no_new_price_extended_with_pca
add_tree_model_metrics('DecisionTree Regressor', 'No New Price', dtr_results_no_new_price)
add_tree_model_metrics('RandomForest Regressor', 'No New Price', rfr_results_no_new_price)
add_tree_model_metrics('GradientBoosting Regressor', 'No New Price', gbr_results_no_new_price)
add_tree_model_metrics('XGBoost Regressor', 'No New Price', xgb_results_no_new_price)
add_tree_model_metrics('LightGBM Regressor', 'No New Price', lgbm_results_no_new_price)
add_tree_model_metrics('CatBoost Regressor', 'No New Price', cbr_results_no_new_price)

# Sort the DataFrame by Test_RMSE in ascending order (lower RMSE is better)
tree_metrics_summary_sorted = tree_metrics_summary.sort_values(by='Test_RMSE', ascending=True).reset_index(drop=True)

# Function to bold the best value for each metric
def highlight_best(s):
    if s.name in ['Train_R2', 'Test_R2']:
        is_best = s == s.max()
    elif s.name in ['Train_RMSE', 'Train_MAE', 'Train_MAPE', 'Test_RMSE', 'Test_MAE', 'Test_MAPE']:
        # Handle potential NaN values by finding min only among non-NaNs
        if s.isnull().all():
            return [''] * len(s)
        valid_min = s.min(skipna=True)
        is_best = (s == valid_min) & (~s.isnull())
    else:
        return [''] * len(s)
    return ['font-weight: bold' if v else '' for v in is_best]

# Display the summary table with highlighting
print("\n--- Tree-based Model Performance Summary (Sorted by Test RMSE, with best metrics bolded) ---")
display(tree_metrics_summary_sorted.style.apply(highlight_best, subset=pd.Index(tree_metrics_summary_sorted.columns).drop(['Model', 'DataFrame'])))
````

### **Hyperparameter Tuning for Top Performing Boosting Models (GridSearchCV)**

Based on the tree_metrics_summary table, the top-performing boosting models are the CatBoostRegressor and the XGBoostRegressor on the df_transformed_with_new_price_extended_with_pca DataFrame.

It's important to note that GridSearchCV can be very resource-intensive, especially with a large number of hyperparameters or when dealing with complex models and a large dataset. Running extensive K-fold Cross-Validation (e.g., more than 3 or 5 folds) or multiple bootstrapping samples (e.g., 100 or more) within the GridSearchCV loop for each combination of hyperparameters would have led to prohibitively long execution times and potentially system timeouts, as was experienced during an attempt at a wider CatBoost tuning.

Therefore, to make progress within practical time limits and available computing resources, a more streamlined approach focusing on train/test performance for initial assessment was adopted. More rigorous validation techniques like full K-fold CV and bootstrapping were reserved for the linear models where the computational burden per fit was lower, and for later stages of model refinement if initial results warranted further investment of resources.

#### **Defining the hyperparameter search spaces for these models:**



````python
from sklearn.model_selection import GridSearchCV

# Define hyperparameter grids for tuning

# CatBoost Regressor
catboost_param_grid = {
    'iterations': [100, 250, 500], # Number of boosting rounds
    'learning_rate': [0.01, 0.05, 0.1],
    'depth': [6, 8, 10] # Depth of the trees
}

# XGBoost Regressor
xgb_param_grid = {
    'n_estimators': [100, 250, 500],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [3, 5, 7]
}

# LightGBM Regressor
lgbm_param_grid = {
    'n_estimators': [100, 250, 500],
    'learning_rate': [0.01, 0.05, 0.1],
    'num_leaves': [31, 63] # Default 31, 2^max_depth for max_depth 5/6
}

print("Hyperparameter grids defined for CatBoost, XGBoost, and LightGBM.")
````

#### **Performing hyperparameter tuning for CatBoost Regressor using GridSearchCV on `df_transformed_with_new_price_extended_with_pca`:**



````python
from catboost import CatBoostRegressor
from sklearn.model_selection import GridSearchCV

# Prepare data for tuning
X_tune = df_transformed_with_new_price_extended_with_pca.drop(columns=["log_Price"])
y_tune = df_transformed_with_new_price_extended_with_pca["log_Price"]

# Convert boolean columns to int type for CatBoost (GridSearchCV might pass data in array format)
for col in X_tune.select_dtypes(include='bool').columns:
    X_tune[col] = X_tune[col].astype(int)

# Identify categorical feature indices for CatBoost
cat_features_indices_tune = [i for i, col in enumerate(X_tune.columns) if X_tune[col].dtype == 'int' and X_tune[col].nunique() < 50 and X_tune[col].min() >= 0 and X_tune[col].max() <= 1] # Assuming one-hot encoded bools are now ints 0/1

print("--- Tuning CatBoost Regressor ---")
cbr_tune = CatBoostRegressor(random_state=42, verbose=False)

grid_search_cbr = GridSearchCV(
    estimator=cbr_tune,
    param_grid=catboost_param_grid,
    cv=3, # Use a smaller CV for tuning to save time
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

grid_search_cbr.fit(X_tune, y_tune, cat_features=cat_features_indices_tune)

best_cbr = grid_search_cbr.best_estimator_
print(f"Best CatBoost parameters: {grid_search_cbr.best_params_}")

cbr_tuned_results = evaluate_tree_model(
    model=best_cbr,
    model_name="CatBoost Regressor (Tuned)",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price",
    cat_features_indices=cat_features_indices_tune
)
````

#### **Performing hyperparameter tuning for the XGBoost Regressor using GridSearchCV on `df_transformed_with_new_price_extended_with_pca`:**



````python
from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV

# Prepare data for tuning
X_tune = df_transformed_with_new_price_extended_with_pca.drop(columns=["log_Price"])
y_tune = df_transformed_with_new_price_extended_with_pca["log_Price"]

# Convert boolean columns to int type for XGBoost
for col in X_tune.select_dtypes(include='bool').columns:
    X_tune[col] = X_tune[col].astype(int)

print("--- Tuning XGBoost Regressor ---")
xgb_tune = XGBRegressor(random_state=42, eval_metric='rmse') # eval_metric added to suppress warning

grid_search_xgb = GridSearchCV(
    estimator=xgb_tune,
    param_grid=xgb_param_grid,
    cv=3, # Use a smaller CV for tuning to save time
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

grid_search_xgb.fit(X_tune, y_tune)

best_xgb = grid_search_xgb.best_estimator_
print(f"Best XGBoost parameters: {grid_search_xgb.best_params_}")

xgb_tuned_results = evaluate_tree_model(
    model=best_xgb,
    model_name="XGBoost Regressor (Tuned)",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price"
)
````

### **Hyperparameter Tuning for Top Performing Boosting Models (Bayesian Optimization)**

Building upon the GridSearchCV results, an alternative hyperparameter tuning method, **Bayesian Optimization with `Optuna`**, is being pursued for the two top-performing boosting models (CatBoost and XGBoost).

#### **Why Bayesian Optimization after GridSearchCV?**

While GridSearchCV systematically explores a predefined grid of hyperparameters, it can be computationally expensive and inefficient for models with large or continuous hyperparameter spaces. Optuna is a hyperparameter optimization framework that, by default, uses a Tree-structured Parzen Estimator (TPE) algorithm, which is a form of Bayesian optimization. This approach intelligently explores the hyperparameter space by building a probabilistic model of the objective function, aiming to find optimal parameters more efficiently than traditional grid search or random search.

**Advantages of Bayesian Optimization:**

1.  **Efficiency**: Unlike GridSearchCV's exhaustive search, Bayesian Optimization intelligently selects the next set of hyperparameters to evaluate based on past results. It balances exploration (trying new, potentially good regions) and exploitation (refining known good regions), leading to faster convergence to better solutions.
2.  **Better Performance**: By guiding the search with a probabilistic model, Bayesian Optimization can often discover superior hyperparameter combinations that might be missed by GridSearchCV's discrete grid, potentially leading to models with genuinely better predictive performance.
3.  **Handling Continuous Spaces**: It is particularly well-suited for continuous hyperparameters or when the optimal values are not expected to fall exactly on a predefined grid point.
4.  **Resource Optimization**: Given the computational intensity observed with comprehensive grid searches, Bayesian Optimization offers a more efficient use of computational resources, allowing for deeper exploration of the hyperparameter space within practical time limits.

By employing Bayesian Optimization with `Optuna` for fine-tuning the best boosting models, the aim is to achieve even higher accuracy and robustness, extracting the maximum potential from these powerful algorithms.

#### **Hyperparameter tuning for CatBoost Regressor using Bayesian Optimization (100 trials) on `df_transformed_with_new_price_extended_with_pca`:**




**Preparing the objective function for Bayesian Optimization:**



````python
import optuna
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from catboost import CatBoostRegressor

# Prepare the data for the objective function
X_optim = df_transformed_with_new_price_extended_with_pca.drop(columns=["log_Price"]).copy()
y_optim = df_transformed_with_new_price_extended_with_pca["log_Price"].copy()

# Convert boolean columns to int type for CatBoost
for col in X_optim.select_dtypes(include='bool').columns:
    X_optim[col] = X_optim[col].astype(int)

# Identify categorical feature indices for CatBoost
cat_features_indices_optim = [i for i, col in enumerate(X_optim.columns) if X_optim[col].dtype == 'int' and X_optim[col].nunique() < 100 and X_optim[col].min() >= 0 and X_optim[col].max() <= 1]

# Split data into training and validation sets for the objective function
X_train_obj, X_val_obj, y_train_obj, y_val_obj = train_test_split(X_optim, y_optim, test_size=0.2, random_state=42)

def objective(trial):
    """
    Objective function for Optuna to optimize CatBoost Regressor hyperparameters.
    It returns the RMSE on a validation set.
    """
    params = {
        'iterations': trial.suggest_int('iterations', 100, 1000),
        'learning_rate': trial.suggest_loguniform('learning_rate', 1e-3, 0.1),
        'depth': trial.suggest_int('depth', 4, 10),
        'l2_leaf_reg': trial.suggest_loguniform('l2_leaf_reg', 1e-2, 10.0),
        'loss_function': 'RMSE',
        'eval_metric': 'RMSE',
        'random_seed': 42,
        'verbose': 0, # Suppress verbose output during trials
        'early_stopping_rounds': 100 # Early stopping to prevent overfitting and speed up training
    }

    model = CatBoostRegressor(**params)

    model.fit(X_train_obj, y_train_obj,
              eval_set=(X_val_obj, y_val_obj),
              cat_features=cat_features_indices_optim,
              early_stopping_rounds=50, # Pass again for fit method
              verbose=0)

    y_pred = model.predict(X_val_obj)
    rmse = np.sqrt(mean_squared_error(y_val_obj, y_pred))

    return rmse

print("Objective function for CatBoost Regressor defined.")
````

**Creating a study object and optimizing the objective function:**


```python
# Create a study object and optimize the objective function
print("--- Starting Optuna optimization for CatBoost Regressor ---")
study_cbr = optuna.create_study(direction='minimize', study_name='CatBoost_Hyperparameter_Optimization')
study_cbr.optimize(objective, n_trials=100) # You can adjust n_trials as needed

print("\n--- Optuna optimization for CatBoost Regressor completed ---")
print(f"Number of finished trials: {len(study_cbr.trials)}")
print(f"Best trial: {study_cbr.best_trial.value}")
print(f"Best parameters: {study_cbr.best_params}")

# Train and evaluate the CatBoost Regressor with the best hyperparameters found by Optuna

cbr_optuna_tuned_results = evaluate_tree_model(
    model=best_cbr_optuna,
    model_name="CatBoost Regressor (Optuna Tuned)",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price",
    cat_features_indices=cat_features_indices_optim # Use the same cat_features_indices as in objective
)
```

**Performing hyperparameter tuning for CatBoost Regressor using Bayesian Optimization (100 trials) on `df_transformed_with_new_price_extended_with_pca`:**

````python
import optuna

# 1. Create an Optuna study object
study_cbr = optuna.create_study(direction='minimize', study_name='CatBoost_Hyperparameter_Optimization')

# 2. Optimize the study_cbr object using the pre-defined objective function for 100 trials
print("Starting Optuna optimization for CatBoost Regressor...")
study_cbr.optimize(objective, n_trials=100, show_progress_bar=True)
print("Optuna optimization for CatBoost Regressor completed.")

# 3. Print the number of finished trials
print(f"Number of finished trials: {len(study_cbr.trials)}")

# 4. Print the best trial value
print(f"Best trial value (RMSE): {study_cbr.best_value:.4f}")

# 5. Print the best parameters found by Optuna
print("Best parameters found by Optuna:")
for key, value in study_cbr.best_params.items():
    print(f"  {key}: {value}")
````

#### **Hyperparameter tuning for XGBoost Regressor using Bayesian Optimization (100 trials) on `df_transformed_with_new_price_extended_with_pca`:**

**Preparing the objective function for Bayesian Optimization:**




````python
import optuna
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from xgboost import XGBRegressor

# Prepare the data for the objective function
X_optim_xgb = df_transformed_with_new_price_extended_with_pca.drop(columns=["log_Price"]).copy()
y_optim_xgb = df_transformed_with_new_price_extended_with_pca["log_Price"].copy()

# Convert boolean columns to int type for XGBoost
for col in X_optim_xgb.select_dtypes(include='bool').columns:
    X_optim_xgb[col] = X_optim_xgb[col].astype(int)

# Split data into training and validation sets for the objective function
X_train_obj_xgb, X_val_obj_xgb, y_train_obj_xgb, y_val_obj_xgb = train_test_split(X_optim_xgb, y_optim_xgb, test_size=0.2, random_state=42)

def objective_xgb(trial):
    """
    Objective function for Optuna to optimize XGBoost Regressor hyperparameters.
    It returns the RMSE on a validation set.
    """
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
        'learning_rate': trial.suggest_loguniform('learning_rate', 1e-3, 0.1),
        'max_depth': trial.suggest_int('max_depth', 3, 10),
        'min_child_weight': trial.suggest_loguniform('min_child_weight', 1e-2, 10),
        'subsample': trial.suggest_float('subsample', 0.5, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.5, 1.0),
        'gamma': trial.suggest_loguniform('gamma', 1e-8, 1.0),
        'reg_alpha': trial.suggest_loguniform('reg_alpha', 1e-8, 1.0),
        'reg_lambda': trial.suggest_loguniform('reg_lambda', 1e-8, 1.0),
        'random_state': 42,
        'tree_method': 'hist', # Use 'hist' for faster training
        'eval_metric': 'rmse' # Evaluation metric
    }

    model = XGBRegressor(**params)

    # Removed 'early_stopping_rounds' from here to fix the TypeError
    model.fit(X_train_obj_xgb, y_train_obj_xgb,
              eval_set=[(X_val_obj_xgb, y_val_obj_xgb)],
              verbose=False)

    y_pred = model.predict(X_val_obj_xgb)
    rmse = np.sqrt(mean_squared_error(y_val_obj_xgb, y_pred))

    return rmse

print("Objective function for XGBoost Regressor defined.")
````

**Performing hyperparameter tuning for XGBoost Regressor using Bayesian Optimization (100 trials) on `df_transformed_with_new_price_extended_with_pca`:**



````python
import optuna

# Create an Optuna study object
study_xgb = optuna.create_study(direction='minimize', study_name='XGBoost_Hyperparameter_Optimization')

# Optimize the study_xgb object using the pre-defined objective_xgb function for 100 trials
print("Starting Optuna optimization for XGBoost Regressor...")
study_xgb.optimize(objective_xgb, n_trials=100, show_progress_bar=True)
print("Optuna optimization for XGBoost Regressor completed.")

# Print the number of finished trials
print(f"Number of finished trials: {len(study_xgb.trials)}")

# Print the best trial value
print(f"Best trial value (RMSE): {study_xgb.best_value:.4f}")

# Print the best parameters found by Optuna
print("Best parameters found by Optuna:")
for key, value in study_xgb.best_params.items():
    print(f"  {key}: {value}")
````

**Evaluating the hyperparameter tuning for XGBoost Regressor using Bayesian Optimization (100 trials) on `df_transformed_with_new_price_extended_with_pca`:**

````python
print("--- Evaluating XGBoost Regressor (Optuna Tuned) ---")
best_xgb_optuna = XGBRegressor(random_state=42, **study_xgb.best_params)

xgb_optuna_tuned_results = evaluate_tree_model(
    model=best_xgb_optuna,
    model_name="XGBoost Regressor (Optuna Tuned)",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price"
)
````

#### **Hyperparameter tuning for CatBoost Regressor using Bayesian Optimization (200 trials) on `df_transformed_with_new_price_extended_with_pca`:**

**Creating Objective Function:**

````python
import optuna
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from catboost import CatBoostRegressor

# Prepare the data for the objective function
X_optim = df_transformed_with_new_price_extended_with_pca.drop(columns=["log_Price"]).copy()
y_optim = df_transformed_with_new_price_extended_with_pca["log_Price"].copy()

# Convert boolean columns to int type for CatBoost
for col in X_optim.select_dtypes(include='bool').columns:
    X_optim[col] = X_optim[col].astype(int)

# Identify categorical feature indices for CatBoost
cat_features_indices_optim = [i for i, col in enumerate(X_optim.columns) if X_optim[col].dtype == 'int' and X_optim[col].nunique() < 200 and X_optim[col].min() >= 0 and X_optim[col].max() <= 1]

# Split data into training and validation sets for the objective function
X_train_obj, X_val_obj, y_train_obj, y_val_obj = train_test_split(X_optim, y_optim, test_size=0.2, random_state=42)

def objective(trial):
    """
    Objective function for Optuna to optimize CatBoost Regressor hyperparameters.
    It returns the RMSE on a validation set.
    """
    params = {
        'iterations': trial.suggest_int('iterations', 100, 1000),
        'learning_rate': trial.suggest_float('learning_rate', 1e-3, 0.1, log=True),
        'depth': trial.suggest_int('depth', 4, 10),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 1e-2, 10.0, log=True),
        'loss_function': 'RMSE',
        'eval_metric': 'RMSE',
        'random_seed': 42,
        'verbose': 0, # Suppress verbose output during trials
        'early_stopping_rounds': 200 # Early stopping to prevent overfitting and speed up training
    }

    model = CatBoostRegressor(**params)

    model.fit(X_train_obj, y_train_obj,
              eval_set=(X_val_obj, y_val_obj),
              cat_features=cat_features_indices_optim,
              early_stopping_rounds=50, # Pass again for fit method
              verbose=0)

    y_pred = model.predict(X_val_obj)
    rmse = np.sqrt(mean_squared_error(y_val_obj, y_pred))

    return rmse

print("Objective function for CatBoost Regressor defined.")
````

**Performing and evaluating hyperparameter tuning for CatBoost Regressor using Bayesian Optimization (200 trials) on `df_transformed_with_new_price_extended_with_pca`:**

````python
import optuna
from catboost import CatBoostRegressor

# Create an Optuna study object with persistent storage for CatBoost
# This ensures trials are not overwritten and can be reloaded/continued.
study_cbr_extended = optuna.create_study(direction='minimize', study_name='CatBoost_Hyperparameter_Optimization_Extended',
                                          storage='sqlite:///catboost_optuna_extended.db', load_if_exists=True)

# Optimize the study object using the pre-defined objective function for 200 trials
print("Starting extended Optuna optimization for CatBoost Regressor (200 trials)...")
study_cbr_extended.optimize(objective, n_trials=200, show_progress_bar=True)
print("Extended Optuna optimization for CatBoost Regressor completed.")

# Print the number of finished trials
print(f"Number of finished trials: {len(study_cbr_extended.trials)}")

# Print the best trial value
print(f"Best trial value (RMSE): {study_cbr_extended.best_value:.4f}")

# Print the best parameters found by Optuna
print("Best parameters found by Optuna:")
for key, value in study_cbr_extended.best_params.items():
    print(f"  {key}: {value}")

# Evaluate the CatBoost Regressor with the best hyperparameters found by this extended Optuna run
print("\n--- Evaluating CatBoost Regressor (Extended Optuna Tuned) ---")
best_cbr_optuna_extended = CatBoostRegressor(random_state=42, verbose=False, **study_cbr_extended.best_params)

cbr_optuna_tuned_extended_results = evaluate_tree_model(
    model=best_cbr_optuna_extended,
    model_name="CatBoost Regressor (Extended Optuna Tuned)",
    df_input=df_transformed_with_new_price_extended_with_pca.copy(),
    target_col="log_Price",
    cat_features_indices=cat_features_indices_optim # Use the same cat_features_indices as in objective
)
````

#### **Summary Table of Tree-based Models and Hyperparameter Tuned Models Performance Metrics:**

````python
tree_metrics_summary_new = pd.DataFrame(columns=[
    'Model', 'DataFrame',
    'Train_R2', 'Train_RMSE', 'Train_MAE', 'Train_MAPE',
    'Test_R2', 'Test_RMSE', 'Test_MAE', 'Test_MAPE'
])

def add_tree_model_metrics_to_new_summary(model_name, dataframe_name, results_dict):
    global tree_metrics_summary_new
    new_row = {
        'Model': model_name,
        'DataFrame': dataframe_name,
        'Train_R2': results_dict['train_metrics']['R2'],
        'Train_RMSE': results_dict['train_metrics']['RMSE'],
        'Train_MAE': results_dict['train_metrics']['MAE'],
        'Train_MAPE': results_dict['train_metrics']['MAPE'],
        'Test_R2': results_dict['test_metrics']['R2'],
        'Test_RMSE': results_dict['test_metrics']['RMSE'],
        'Test_MAE': results_dict['test_metrics']['MAE'],
        'Test_MAPE': results_dict['test_metrics']['MAPE']
    }
    tree_metrics_summary_new = pd.concat([tree_metrics_summary_new, pd.DataFrame([new_row])], ignore_index=True)

# Add metrics for models evaluated on df_transformed_with_new_price_extended_with_pca
add_tree_model_metrics_to_new_summary('DecisionTree Regressor', 'With New Price', dtr_results_with_new_price)
add_tree_model_metrics_to_new_summary('RandomForest Regressor', 'With New Price', rfr_results_with_new_price)
add_tree_model_metrics_to_new_summary('GradientBoosting Regressor', 'With New Price', gbr_results_with_new_price)
add_tree_model_metrics_to_new_summary('XGBoost Regressor', 'With New Price', xgb_results_with_new_price)
add_tree_model_metrics_to_new_summary('LightGBM Regressor', 'With New Price', lgbm_results_with_new_price)
add_tree_model_metrics_to_new_summary('CatBoost Regressor', 'With New Price', cbr_results_with_new_price)

# Add metrics for models evaluated on df_transformed_no_new_price_extended_with_pca
add_tree_model_metrics_to_new_summary('DecisionTree Regressor', 'No New Price', dtr_results_no_new_price)
add_tree_model_metrics_to_new_summary('RandomForest Regressor', 'No New Price', rfr_results_no_new_price)
add_tree_model_metrics_to_new_summary('GradientBoosting Regressor', 'No New Price', gbr_results_no_new_price)
add_tree_model_metrics_to_new_summary('XGBoost Regressor', 'No New Price', xgb_results_no_new_price)
add_tree_model_metrics_to_new_summary('LightGBM Regressor', 'No New Price', lgbm_results_no_new_price)
add_tree_model_metrics_to_new_summary('CatBoost Regressor', 'No New Price', cbr_results_no_new_price)

# Add tuned models
add_tree_model_metrics_to_new_summary('CatBoost Regressor (GridSearchCV Tuned)', 'With New Price', cbr_tuned_results)
add_tree_model_metrics_to_new_summary('XGBoost Regressor (GridSearchCV Tuned)', 'With New Price', xgb_tuned_results)
add_tree_model_metrics_to_new_summary('CatBoost Regressor (Optuna Tuned)', 'With New Price', cbr_optuna_tuned_extended_results)
add_tree_model_metrics_to_new_summary('XGBoost Regressor (Optuna Tuned)', 'With New Price', xgb_optuna_tuned_extended_results)

# Sort the DataFrame by Test_RMSE in ascending order (lower RMSE is better)
tree_metrics_summary_sorted_new = tree_metrics_summary_new.sort_values(by='Test_RMSE', ascending=True).reset_index(drop=True)

# Function to bold the best value for each metric (re-defined for clarity in new cell)
def highlight_best_new(s):
    if s.name in ['Train_R2', 'Test_R2']:
        is_best = s == s.max()
    elif s.name in ['Train_RMSE', 'Train_MAE', 'Train_MAPE', 'Test_RMSE', 'Test_MAE', 'Test_MAPE']:
        if s.isnull().all():
            return [''] * len(s)
        valid_min = s.min(skipna=True)
        is_best = (s == valid_min) & (~s.isnull())
    else:
        return [''] * len(s)
    return ['font-weight: bold' if v else '' for v in is_best]

# Display the summary table with highlighting
print("\n--- Tree-based Model Performance Summary (Sorted by Test RMSE, with best metrics bolded) ---")
display(tree_metrics_summary_sorted_new.style.apply(highlight_best_new, subset=pd.Index(tree_metrics_summary_sorted_new.columns).drop(['Model', 'DataFrame'])))
````

### **Implementing SHapley Additive exPlanations (SHAP) for Best Tree-based Model (`XGBoost Regressor (Optuna Tuned))`**





````python
# Install shap if not already installed
!pip install shap

import shap
import matplotlib.pyplot as plt

# Ensure the best model and data are available
# The best model was stored in `best_xgb_optuna_extended`
# The data used for training the objective function was `X_train_obj_xgb` and `y_train_obj_xgb`

# Prepare the data for SHAP analysis. Using X_train_obj_xgb as it's the training data for the final tuned model.
X_shap = X_train_obj_xgb.copy()
y_shap = y_train_obj_xgb.copy()

# Create a SHAP explainer object
# Using TreeExplainer for tree-based models like XGBoost
explainer = shap.TreeExplainer(best_xgb_optuna_extended)

# Calculate SHAP values for the training data
# This might take a moment depending on the dataset size and model complexity
print("Calculating SHAP values...")
shap_values = explainer.shap_values(X_shap)
print("SHAP values calculated.")

# --- Visualize SHAP Summary Plots ---

# 1. SHAP Summary Plot (beeswarm - global feature importance and impact direction)
print("\nGenerating SHAP summary plot (beeswarm)...\n")
shap.summary_plot(shap_values, X_shap, show=False)
plt.title('SHAP Feature Importance (Beeswarm Plot)')
plt.tight_layout()
plt.show()

# 2. SHAP Bar Plot (global average feature importance magnitude)
print("\nGenerating SHAP bar plot (mean absolute SHAP value)...\n")
shap.summary_plot(shap_values, X_shap, plot_type="bar", show=False)
plt.title('SHAP Feature Importance (Bar Plot)')
plt.tight_layout()
plt.show()

# --- Visualize SHAP Dependence Plots (for a few key features) ---
# Selecting a few top features based on the summary plot for dependence plots
# To do this programmatically, we can get the feature order from the summary plot
feature_importance = pd.DataFrame({
    'feature': X_shap.columns,
    'shap_abs': np.abs(shap_values).mean(axis=0)
}).sort_values(by='shap_abs', ascending=False)

top_features = feature_importance['feature'].head(5).tolist()

print("\nGenerating SHAP dependence plots for top features...\n")
for feature in top_features:
    # Only plot if the feature has varying SHAP values and is not a constant
    if X_shap[feature].nunique() > 1 and np.std(shap_values[:, X_shap.columns.get_loc(feature)]) > 0:
        shap.dependence_plot(feature, shap_values, X_shap, interaction_index="auto", show=False)
        plt.suptitle(f'SHAP Dependence Plot for {feature}', y=1.02)
        plt.tight_layout()
        plt.show()
    else:
        print(f"Skipping dependence plot for {feature} due to constant or zero-variance values.")

# --- Visualize SHAP Force Plots (for individual predictions) ---
# Select a few random samples from the test set for individual explanations
print("\nGenerating SHAP force plots for individual predictions...\n")

# Using X_test from the last evaluation run of the XGBoost model
# Assuming X_test_scaled_df was created and contains the correctly scaled test features
# If X_test_scaled_df is not available, we need to regenerate X_test from df_input

# Re-split data to get X_test if not readily available in the kernel from a previous run
current_df = df_transformed_with_new_price_extended_with_pca.copy()
initial_rows_count = len(current_df)
current_df.replace([np.inf, -np.inf], np.nan, inplace=True)
current_df.dropna(inplace=True)
if len(current_df) < initial_rows_count:
    print(f"Warning: Dropped {initial_rows_count - len(current_df)} rows from the input DataFrame due to inf/nan values when preparing for SHAP force plot.")

X_temp_all = current_df.drop(columns=["log_Price"])
y_temp_all = current_df["log_Price"]

for col in X_temp_all.select_dtypes(include='bool').columns:
    X_temp_all[col] = X_temp_all[col].astype(int)

_, X_test_force_plot, _, _ = train_test_split(X_temp_all, y_temp_all, test_size=0.2, random_state=42)

# Generate a new explainer and shap_values for the test set for clarity in force plots
explainer_test = shap.TreeExplainer(best_xgb_optuna_extended)
shap_values_test = explainer_test.shap_values(X_test_force_plot)

# Randomly select 3 samples from the test set
sample_indices = X_test_force_plot.sample(n=3, random_state=42).index

for i, idx in enumerate(sample_indices):
    print(f"Explanation for Test Sample {i+1} (Original Index: {idx}):")
    # Use init_js() to ensure JS is loaded for interactive plots in notebooks
    shap.initjs()
    # Display force plot for the individual prediction
    display(shap.force_plot(explainer_test.expected_value, shap_values_test[X_test_force_plot.index.get_loc(idx)], X_test_force_plot.loc[idx]))

###print("\nSHAP analysis complete.")
````

This indicates that it has learned generalizable patterns without significantly overfitting, likely due to effective hyperparameter tuning via Bayesian Optimization.

### **5. Handling of Data Complexities:**

*   **OLS Strengths**: OLS with transformations (`log_Price`, `log_Kilometers_Driven`, `log_Engine`, `log_Power`, interaction terms) can capture some non-linear relationships. The use of PCA for `log_Power` and `log_Engine` addresses multicollinearity between these specific features. Robust standard errors directly address heteroskedasticity. However, it implicitly assumes linear relationships after transformations, and struggles with severe multicollinearity among one-hot encoded categorical features (as seen by infinite VIFs in some cases).
*   **XGBoost Strengths**: XGBoost inherently handles non-linear relationships and feature interactions very well due to its tree-based structure. It is also more robust to multicollinearity and outliers compared to OLS. It does not make assumptions about the distribution of errors or homoskedasticity, making it suitable for complex, real-world datasets where these assumptions are often violated.
*   **XGBoost Weakness**: While powerful, XGBoost does not directly provide the clear statistical diagnostics (like p-values for individual features or formal tests for heteroskedasticity) that OLS does, making it harder to formally evaluate the significance of individual predictors in the same statistical inference framework.

### **6. Computational Considerations:**

*   **OLS**: Generally faster to train, especially on moderate datasets. The primary computational cost comes from matrix inversions, but it's relatively efficient.
*   **XGBoost**: More computationally intensive to train, particularly during hyperparameter tuning (even with efficient methods like Optuna), as it builds an ensemble of many decision trees sequentially. Prediction is generally fast once trained.

### **Conclusion and Recommendation:**

For this project, given the primary objective to **accurately estimate the market price of a used car**, the **`XGBoost Regressor (Optuna Tuned)` model is the superior choice.** While the `OLS_Robust_Best` model offers excellent interpretability and strong performance, XGBoost achieves significantly higher predictive accuracy (higher R-squared and lower RMSE, MAE, and MAPE) on unseen data. This indicates its better ability to capture the complex, non-linear relationships and interactions that drive used car prices.

**Recommendation:** Deploy the `XGBoost Regressor (Optuna Tuned)` model for its top-tier predictive accuracy. However, the insights derived from the OLS model's coefficients regarding the direction and general magnitude of feature impacts (e.g., `log_New_price` is a strong positive driver, `Car_Age` a negative one) remain valuable for business understanding and guiding future feature engineering efforts.

# **Final Conclusion and Recommendations for Cars4U Used Price Prediction**

### **Problem and Project Objective:**
Cars4U faced significant challenges in the highly fragmented and volatile Indian used car market due to inaccurate pricing, leading to both overpricing (reducing sales velocity) and underpricing (resulting in revenue loss). The primary objective of this project was to develop a robust, data-driven predictive model to accurately estimate the market price of a used car based on its attributes.

### **Key Findings from Data Analysis and Preprocessing:**

1.  **Target Variable (`Price`) Handling:** Initial data inspection revealed significant missing values in the target `Price` variable. Rows with missing prices were dropped to ensure model integrity, reducing the dataset from 7253 to 6019 observations. The target variable was successfully log-transformed into `log_Price` to address its severe right-skewness, improving its suitability for regression models.
2.  **`New_price` - A Powerful, Yet Missing, Feature:** The `New_price` column had a very high proportion of missing values (6247 out of 7253 initially). Despite a multi-phase imputation strategy using grouped medians (`Brand`, `Year`, `Fuel_Type`, `Transmission`, `Engine`, `Power`, `Seats`), approximately 55% of `New_price` values remained missing. However, subsequent modeling unequivocally demonstrated `log_New_price` to be the **single most influential predictor** of used car prices.
3.  **Numerical Feature Transformations:** Highly skewed numerical features like `Kilometers_Driven`, `Engine`, and `Power` were successfully log-transformed into `log_Kilometers_Driven`, `log_Engine`, and `log_Power` to normalize their distributions and enhance model performance. A notable outlier in `Kilometers_Driven` (6.5 million km) was removed, and anomalous 0.0 `Mileage` entries were handled.
4.  **Feature Engineering for Enhanced Predictive Power:**
    *   **`Car_Age`:** Created from `Year` (2019 - `Year`) to provide a direct measure of vehicle age, reflecting depreciation.
    *   **Interaction Terms:** Crucial interaction terms like `Mileage_x_Car_Age_Interaction`, `log_Kilometers_Driven_x_Car_Age_Interaction`, and `log_New_price_x_Car_Age_Interaction` were engineered to capture combined effects (e.g., how age impacts depreciation differently for luxury versus economy cars, or the combined effect of mileage and age).
    *   **`Kilometers_per_Year`:** Derived from `Kilometers_Driven` / `Car_Age` to represent average annual usage.
    *   **Multicollinearity Management:** High correlation between `log_Power` and `log_Engine` was addressed by applying Principal Component Analysis (PCA), creating `Power_Engine_PCA` to capture their combined variance efficiently.
5.  **Categorical Feature Refinements:**
    *   `Brand` was extracted from `Name`, allowing for brand-specific pricing dynamics.
    *   `Fuel_Type_Grouped` was created (combining 'CNG'/'LPG' into 'Natural_Gas') and `Owner_Type_Revised` was created (grouping 'Third'/'Fourth & Above' into 'Multiple') to manage sparsity and improve model learning for less frequent categories.
    *   `Seats` was treated as a categorical variable, recognizing its non-linear relationship with price.
    *   All refined categorical features were one-hot encoded for model input.

### **Key Takeaways from Model Building and Evaluation:**

1.  **Superiority of Tree-based Models:** Ensemble tree-based models, particularly Gradient Boosting Machines like XGBoost and CatBoost, significantly outperformed traditional Ordinary Least Squares (OLS) regression models in terms of predictive accuracy.
2.  **`log_New_price` as the Ultimate Driver:** The inclusion of `log_New_price` and its engineered interactions dramatically improved model performance across all model types. Models incorporating these features (though operating on a smaller dataset of 2644 observations due to `New_price` missingness) consistently achieved higher R-squared and lower error metrics compared to models that excluded them.
3.  **Best Performing Model:** The **`XGBoost Regressor (Optuna Tuned)` model**, trained on the `df_transformed_with_new_price_extended_with_pca` feature set, emerged as the most accurate. It achieved an exceptional **Test R-squared of 0.9704**, a low **Test RMSE of 0.1387**, and an impressive **Test MAPE of 6.19%**. This indicates that, on average, the model's price predictions deviate by only ~6.2% from actual prices.
4.  **Robust OLS Performance:** While outperformed by XGBoost, the `OLS_Robust_Best` model still delivered strong results (Test R-squared 0.9528, Test MAPE 8.15%) and offered direct interpretability of coefficients with robust standard errors, which is valuable for understanding general relationships, despite persistent heteroskedasticity and multicollinearity issues for one-hot encoded features.
5.  **Interpretability through SHAP:** SHAP analysis effectively demystified the `XGBoost` model, confirming that it logically relies on key economic factors. It showed that `log_New_price` and `Power_Engine_PCA` are the most influential global predictors, followed by depreciation drivers (`Car_Age` and `Kilometers_Driven` interactions), and features representing brand, location, and transmission type.

### **Final Proposed Model Specifications:**
*   **Model Type:** XGBoost Regressor (Optuna Tuned)
*   **Target Variable:** `log_Price` (log-transformed used car price)
*   **Feature Set:** `df_transformed_with_new_price_extended_with_pca`, encompassing log-transformed core numerical features, engineered terms capturing non-linearities and interactions (`Car_Age`, `Car_Age_Squared`, `Mileage`, `Kilometers_per_Year`, `Mileage_x_Car_Age_Interaction`, `log_Kilometers_Driven_x_Car_Age_Interaction`, `log_New_price_x_Car_Age_Interaction`), the `Power_Engine_PCA` component, a `Is_Premium_New_Price` binary indicator, and one-hot encoded categorical features (`Location`, `Fuel_Type_Grouped`, `Transmission`, `Owner_Type_Revised`, `Brand`, `Seats`).

### **Key Next Steps and Recommendations for Cars4U:**

1.  **Deploy the XGBoost Model:** Implement the tuned XGBoost Regressor into Cars4U's operational systems via a robust MLOps pipeline. This will enable real-time, data-driven pricing decisions, directly addressing underpricing and overpricing.
2.  **Prioritize `New_price` Data Acquisition:** Initiate and sustain efforts to collect complete and accurate `New_price` data for all car models. Given its profound impact on predictive accuracy, this is the most critical data strategy to ensure the model's long-term effectiveness.
3.  **Continuous Monitoring & Retraining:** Establish a system for monitoring model performance and data drift. Regular retraining with new market data will ensure the model remains adaptive and accurate in a dynamic market, maintaining its predictive edge.
4.  **Leverage SHAP for Business Insights:** Utilize SHAP explanations to provide sales agents and stakeholders with transparent, interpretable reasons behind each car's predicted price, fostering trust and aiding negotiation processes.
5.  **Explore Advanced Strategies:** Investigate building segment-specific models (e.g., for luxury vs. economy cars) or incorporating time-series features (e.g., economic indicators) to further enhance model accuracy and robustness against market fluctuations. This can be phased in after initial deployment and validation of the current model.

This comprehensive approach, anchored by the high-performing XGBoost model and supported by clear data strategies, will enable Cars4U to achieve competitive, transparent, and profitable pricing decisions.
